## 1. O que é o Pandas e suas Estruturas Fundamentais

O **Pandas** foi construído sobre o NumPy, unindo a velocidade dos arrays em C à flexibilidade de tabelas com rótulos e tipos heterogêneos por coluna.

### 🧱 As Duas Estruturas Centrais do Pandas:

1. **`Series` (Unidimensional)**:
   * É uma coluna única de dados com um **índice associado** (*Index*).
   * Pense na `Series` como um vetor do NumPy turbinado com rótulos.
2. **`DataFrame` (Bidimensional)**:
   * É uma **tabela completa** composta por várias colunas (`Series`) que compartilham o mesmo índice de linhas.
   * Possui nomes de colunas (*Columns*) e rótulos de linhas (*Index*).

```
         DataFrame (Tabela Bidimensional)
         Colunas: 'Nome'       'Idade'     'Cidade'
Linha 0 --------> "Ana"          25        "São Paulo"
Linha 1 --------> "Carlos"       32        "Rio de Janeiro"
Linha 2 --------> "Mariana"      28        "Belo Horizonte"
                   │
                   └─── Cada coluna individual é uma Series!
```

## 4. Seleção, Fatiamento e Filtragem de Dados no Pandas

O Pandas oferece duas propriedades fundamentais para recortar dados: **`.loc[]`** e **`.iloc[]`**.

### 📌 Diferença Fundamental entre `.loc[]` e `.iloc[]`:

| Propriedade | Tipo de Acesso | Sintaxe | Inclusão do Fim do Intervalo |
| :--- | :--- | :--- | :--- |
| **`.loc[linhas, colunas]`** | **Por Rótulo / Label** | `df.loc[0:2, ['Nome', 'Preco']]` | **Inclusivo**: Inclui o rótulo final (`0:2` pega 0, 1 e 2). |
| **`.iloc[linhas, colunas]`** | **Por Posição Inteira** (estilo Python/NumPy) | `df.iloc[0:2, 0:3]` | **Exclusivo**: Exclui o índice final (`0:2` pega apenas 0 e 1). |

---

### 🎭 Filtragem Condicional com Máscaras Booleanas:
Podemos filtrar linhas passando expressões lógicas entre colchetes:
* **Condição Simples**: `df[df["Preco"] > 200]`
* **Operadores Lógicos no Pandas** (atenção: em Pandas usamos `&`, `|` e `~` com parênteses obrigatórios em cada condição):
  * `&` (*AND*): `df[(df["Preco"] > 100) & (df["Estoque"] > 10)]`
  * `|` (*OR*): `df[(df["Categoria"] == "Áudio") | (df["Preco"] < 150)]`
  * `~` (*NOT*): `df[~(df["Categoria"] == "Monitores")]`
  * `.isin(lista)`: `df[df["Categoria"].isin(["Periféricos", "Áudio"])]`


````python
# Exemplos de uso do .loc e .iloc
print("--- Usando .loc (Rótulo) ---")
print(df_produtos.loc[0:2, ["Produto", "Preco"]])

print("\n--- Usando .iloc (Posição Numérica 0 a 2 e colunas 0 a 2) ---")
print(df_produtos.iloc[0:2, 0:2])

print("\n--- Filtrando produtos com Preço >= 200 E Estoque >= 10 ---")
filtro_composto = (df_produtos["Preco"] >= 200) & (df_produtos["Estoque"] >= 10)
df_produtos[filtro_composto]
````

## 5. Limpeza de Dados e Tratamento de Valores Ausentes (`NaN`)

Em bases de dados reais, é extremamente comum encontrar dados ausentes (*Missing Values*), representados pelo Pandas como **`NaN`** (*Not a Number*) ou `None`.

### 🛠️ Principais Ferramentas de Tratamento de `NaN`:
* **`df.isna().sum()`**: Conta a quantidade de valores nulos existentes em cada coluna.
* **`df.dropna()`**: Descarta linhas que contenham pelo menos um valor nulo (`subset=['coluna']` descarta apenas se o nulo estiver nas colunas especificadas).
* **`df.fillna(valor)`**: Substitui os valores nulos por um valor constante ou calculado (ex: a média ou mediana da coluna: `df['idade'].fillna(df['idade'].mean())`).


````python
# Exemplo de DataFrame com valores ausentes (NaN)
tabela_incompleta = pd.DataFrame({
    "Cliente": ["João", "Maria", "Pedro", "Carla", "Lucas"],
    "Idade": [35, np.nan, 28, 42, np.nan],
    "Gasto_RS": [120.0, 350.0, np.nan, 210.0, 80.0]
})

print("Tabela com Valores Nulos:\n", tabela_incompleta)
print("\nContagem de Nulos por Coluna:\n", tabela_incompleta.isna().sum())

# Preenchendo Idades ausentes com a média e Gastos com zero
idade_media = tabela_incompleta["Idade"].mean()
tabela_tratada = tabela_incompleta.copy()
tabela_tratada["Idade"] = tabela_tratada["Idade"].fillna(idade_media)
tabela_tratada["Gasto_RS"] = tabela_tratada["Gasto_RS"].fillna(0.0)

print("\nTabela após Tratamento (.fillna()):\n", tabela_tratada)
````

## 6. Agrupamento e Agregações com `groupby`

O método **`df.groupby('coluna')`** é uma das ferramentas mais poderosas do Pandas. Ele implementa o paradigma clássico **Split-Apply-Combine**:

1. **Split (Dividir)**: Separa os dados em grupos com base nos valores únicos de uma ou mais colunas.
2. **Apply (Aplicar)**: Executa uma função de agregação matemática sobre cada grupo (ex: soma, média, contagem).
3. **Combine (Combinar)**: Junta os resultados em uma nova tabela estruturada.

```
Tabela Original:                    1. Split (por Categoria):
| Produto | Categoria | Venda |     ┌─────────────┐
| Teclado | Perif     | 200   | ──> │ Periféricos │ -> Apply: sum(200, 100) = 300
| Mouse   | Perif     | 100   |     └─────────────┘
| Monitor | Telas     | 800   |     ┌─────────────┐
                                    │ Telas       │ -> Apply: sum(800) = 800
                                    └─────────────┘
                                    3. Combine:
                                    | Categoria | Venda |
                                    | Perif     | 300   |
                                    | Telas     | 800   |
```
