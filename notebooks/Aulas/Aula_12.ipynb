## 1. Erros de Sintaxe vs Exceções em Tempo de Execução

No Python, os erros são divididos em duas categorias fundamentais:

1. **Erros de Sintaxe (`SyntaxError`)**:
   - Ocorrem quando o interpretador do Python analisa o código (*parsing*) e detecta que a gramática da linguagem foi violada (ex: parênteses abertos e não fechados, dois pontos `:` esquecidos, palavras-chave digitadas incorretamente).
   - O programa **nem chega a iniciar** a execução.

2. **Exceções em Tempo de Execução (*Runtime Exceptions*)**:
   - O código possui sintaxe perfeitamente válida, mas **durante a execução**, ocorre uma situação anômala que o Python não consegue resolver (ex: dividir um número por zero, converter `"abc"` para `int`, acessar um índice inexistente em uma lista).
   - Se a exceção não for tratada por um bloco `try / except`, o programa é interrompido imediatamente com um *Traceback*.

---

### 📋 Principais Exceções Nativas do Python:

| Exceção | Quando Ocorre? | Exemplo Típico |
| :--- | :--- | :--- |
| **`ZeroDivisionError`** | Tentativa de divisão ou módulo por zero. | `10 / 0` ou `7 % 0` |
| **`ValueError`** | Tipo do dado está correto, mas o valor é inválido. | `int("cinco")` ou `float("NaN_invalido")` |
| **`TypeError`** | Operação aplicada a um tipo de dado incompatível. | `'texto' + 5` ou `len(100)` |
| **`IndexError`** | Acesso a um índice fora dos limites de uma sequência. | `lista[10]` em uma lista de 3 elementos |
| **`KeyError`** | Busca por uma chave inexistente em um dicionário. | `dados['telefone']` se a chave não existir |
| **`FileNotFoundError`** | Tentativa de abrir um arquivo que não existe no disco. | `open('relatorio_2020.csv')` |
| **`AttributeError`** | Tentativa de acessar um método/atributo inexistente. | `'texto'.append(10)` |

## 2. O Bloco `try` e `except`

O bloco `try / except` permite interceptar exceções e definir um plano de contingência para que o programa continue sua execução normalmente.

### 🔄 Fluxo de Execução:
1. O Python tenta executar as instruções dentro do bloco **`try`**.
2. Se **nenhuma exceção** ocorrer, o bloco `except` é completamente ignorado.
3. Se **ocorrer uma exceção**, a execução do bloco `try` é imediatamente pausada e o controle é transferido para o bloco **`except`** correspondente.

### ⚠️ Regra de Ouro: Capture Exceções Específicas!
Nunca utilize `except:` sem especificar o tipo de erro (*bare except*). Isso oculta bugs reais (inclusive interrupções de teclado como `KeyboardInterrupt` e erros de digitação de variáveis).


````python
from typing import Optional
# Exemplo Prático: Entrada Segura de Dados do Usuário
def converter_idade_segura(entrada_texto: str) -> Optional[int]:
    try:
        idade = int(entrada_texto)
        print(f"Idade convertida com sucesso: {idade} anos.")
        return idade
    except ValueError as erro:
        print(f"Erro de conversão: '{entrada_texto}' não é um número inteiro válido!")
        print(f"Detalhe técnico da exceção: {erro}")
        return None

# Testando com valores válidos e inválidos
val1 = converter_idade_segura("25")
print(f"Retorno 1: {val1}\n")

val2 = converter_idade_segura("vinte e cinco")
print(f"Retorno 2: {val2}")
````

## 3. Múltiplos Blocos `except` e Hierarquia de Exceções

Uma mesma operação pode falhar por diferentes motivos. Você pode encadear múltiplos blocos `except` para tratar cada falha de maneira personalizada.

Além disso, podemos agrupar erros semelhantes em uma mesma tupla: `except (TypeError, ValueError) as e:`.

### 🌳 Árvore de Hierarquia das Principais Exceções

Em Python, todas as exceções são classes organizadas em uma hierarquia de herança a partir de `BaseException`:

```text
BaseException (Classe raiz do Python)
 ├── SystemExit (Encerramento do interpretador via sys.exit)
 ├── KeyboardInterrupt (Interrupção pelo usuário via Ctrl+C)
 └── Exception (Superclasse de quase todos os erros em tempo de execução)
      ├── ArithmeticError
      │    ├── ZeroDivisionError (Divisão por zero)
      │    └── OverflowError (Número excede capacidade)
      ├── LookupError
      │    ├── IndexError (Índice inexistente em lista/tupla/string)
      │    └── KeyError (Chave inexistente em dicionário)
      ├── ValueError (Tipo correto, mas valor/formato inadequado)
      ├── TypeError (Operação aplicada a tipo incompatível)
      ├── OSError (Erros de sistema operacional)
      │    └── FileNotFoundError (Arquivo ou diretório não encontrado)
      └── MinhasExcecoesPersonalizadas (Herde sempre de Exception!)
```

> 💡 **Regra de Ouro (Ordem de Captura)**: O Python avalia os blocos `except` sequencialmente de **cima para baixo**. Capture sempre as exceções mais **específicas** (folhas da árvore) primeiro e as mais **genéricas** (`Exception`) por último. Se capturar `Exception` no primeiro bloco, nenhuma exceção filha subsequente será executada!


````python
from typing import Optional
# Exemplo Prático: Consulta e Cálculo Seguro em Dicionário de Preços
tabela_precos = {"arroz": 25.50, "feijao": 8.90, "leite": 5.20}

def calcular_valor_item(tabela: dict, produto: str, quantidade_str: str) -> Optional[float]:
    try:
        preco_unitario = tabela[produto]          # Pode gerar KeyError
        quantidade = float(quantidade_str)         # Pode gerar ValueError
        total = preco_unitario * quantidade        # Pode gerar TypeError se tabela tiver tipo errado
        return round(total, 2)
    except KeyError:
        print(f"❌ Erro de Catálogo: Produto '{produto}' não encontrado na tabela.")
        return None
    except ValueError:
        print(f"❌ Erro de Quantidade: '{quantidade_str}' não é uma quantidade numérica válida.")
        return None
    except Exception as e:
        print(f"❌ Erro Inesperado: {type(e).__name__} -> {e}")
        return None

# Testando diferentes cenários de erro
print("Cenário 1 (Sucesso):", calcular_valor_item(tabela_precos, "arroz", "3"))
print("Cenário 2 (Item Inexistente):", calcular_valor_item(tabela_precos, "chocolate", "2"))
print("Cenário 3 (Quantidade Inválida):", calcular_valor_item(tabela_precos, "leite", "duas caixas"))
````

## 4. O Fluxo Completo: `try`, `except`, `else` e `finally`

O Python oferece uma estrutura completa e elegante de fluxo de controle para tratamento de exceções com quatro blocos integrados:

```text
┌─────────────────────────────────────────────────────────────┐
│  try:                                                       │
│  Código principal / operação sujeita a erros               │
└──────────────┬───────────────────────────────┬──────────────┘
               │ (Se ocorreu erro)             │ (Se NÃO houve erro)
               ▼                               ▼
┌─────────────────────────────┐ ┌─────────────────────────────┐
│  except:                    │ │  else:                      │
│  Tratamento do erro         │ │  Executa exclusivamente     │
│  e recuperação da falha     │ │  quando deu tudo certo      │
└──────────────┬──────────────┘ └──────────────┬──────────────┘
               │                               │
               └───────────────┬───────────────┘
                               ▼
┌─────────────────────────────────────────────────────────────┐
│  finally:                                                   │
│  Executa SEMPRE (incondicionalmente, com erro ou sucesso).  │
│  Ideal para limpeza, fechamento de arquivos e conexões.    │
└─────────────────────────────────────────────────────────────┘
```

### 🎯 Para que serve o `finally`?
O bloco `finally` é essencial para **liberar recursos do sistema** (*clean-up*), garantindo execução mesmo se houver um `return` prévio ou uma exceção não tratada:
* Fechar arquivos abertos (`arquivo.close()`).
* Encerrar conexões com bancos de dados ou gateways de pagamento.
* Liberar travas de concorrência (*locks* / *mutexes*).
* Fechar conexões de rede / *sockets*.


````python
# Exemplo Prático: Simulação de Transação com Conexão de Banco de Dados
def processar_pagamento_seguro(valor_centavos: int) -> bool:
    print("🔌 [1] Abrindo conexão com o Gateway de Pagamento...")
    conexao_aberta = True
    
    try:
        if valor_centavos <= 0:
            raise ValueError("O valor do pagamento deve ser estritamente positivo!")
        print(f"💳 [2] Cobrando R$ {valor_centavos / 100:.2f} com sucesso...")
        return True
    except ValueError as err:
        print(f"⚠️ [3] Falha na validação da transação: {err}")
        return False
    else:
        print("✅ [4] Transação aprovada e registrada no livro contábil.")
    finally:
        # Este bloco executa SEMPRE, garantindo o fechamento da conexão
        if conexao_aberta:
            print("🔒 [5] Fechando conexão com o Gateway com segurança (Clean-up).")

print("--- TESTE 1: PAGAMENTO VÁLIDO ---")
processar_pagamento_seguro(15000)

print("\n--- TESTE 2: PAGAMENTO INVÁLIDO ---")
processar_pagamento_seguro(-500)
````

## 5. Disparando Exceções com `raise`

Nem todo erro é causado por um problema técnico do Python. Frequentemente, queremos sinalizar que uma **regra de negócio** foi violada (ex: uma idade negativa, um saque maior que o saldo, ou um código de produto inválido).

Utilizamos a palavra-chave **`raise`** seguida da instância da exceção desejada:

```python
if saldo < valor_saque:
    raise ValueError(f"Saldo insuficiente! Saldo atual: {saldo}, Saque solicitado: {valor_saque}")
```

Você também pode usar `raise` sem argumentos dentro de um bloco `except` para re-lançar a mesma exceção após registrar um log.


````python
# Exemplo Prático: Validador de Regras de Negócio com raise
def cadastrar_usuario_sistema(nome: str, idade: int) -> dict:
    if not isinstance(nome, str) or len(nome.strip()) == 0:
        raise ValueError("O nome do usuário não pode ser vazio ou nulo.")
    
    if idade < 18:
        raise PermissionError(f"Usuário '{nome}' tem {idade} anos. Cadastro restrito a maiores de 18 anos.")
    
    if idade > 130:
        raise ValueError(f"Idade informada ({idade}) é biologicamente implausível.")
        
    return {"status": "SUCESSO", "nome": nome.strip(), "idade": idade}

# Chamando a função com tratamento externo
usuarios_teste = [("Mariana", 22), ("Pedro", 15), ("", 30)]

for nome, idade in usuarios_teste:
    try:
        resultado = cadastrar_usuario_sistema(nome, idade)
        print(f"Cadastro concluído: {resultado}")
    except (ValueError, PermissionError) as e:
        print(f"Falha no cadastro de ({nome}, {idade}): {type(e).__name__} -> {e}")
````

## 6. Criando Exceções Personalizadas (*Custom Exceptions*)

Em sistemas corporativos de grande porte, as exceções padrão do Python (`ValueError`, `RuntimeError`) podem ser genéricas demais.

Podemos criar **nossas próprias classes de exceção** herdando diretamente da classe base `Exception`. Isso permite:
1. Criar erros com nomes claros e específicos para o domínio do problema (ex: `SaldoInsuficienteError`, `TokenExpiradoError`, `ItemEsgotadoError`).
2. Armazenar atributos e dados adicionais de contexto dentro da própria exceção.
3. Capturar exclusivamente erros do nosso módulo sem interferir em erros de bibliotecas de terceiros.


````python
# Exemplo Prático: Sistema de Exceções Customizadas para Games/E-commerce
class SaldoInsuficienteError(Exception):
    """Exceção disparada quando o jogador tenta comprar um item sem moedas suficientes."""
    def __init__(self, saldo_atual: int, preco_item: int, nome_item: str):
        self.saldo_atual = saldo_atual
        self.preco_item = preco_item
        self.nome_item = nome_item
        self.deficit = preco_item - saldo_atual
        super().__init__(
            f"Moedas insuficientes para comprar '{nome_item}'. "
            f"Preço: {preco_item} | Saldo: {saldo_atual} | Faltam: {self.deficit} moedas."
        )

class InventarioCheioError(Exception):
    """Exceção disparada quando a mochila do jogador atinge a capacidade máxima."""
    pass

# Função que utiliza as exceções customizadas
def comprar_item_rpg(jogador: dict, nome_item: str, preco: int, slots_ocupados: int, max_slots: int = 5):
    if slots_ocupados >= max_slots:
        raise InventarioCheioError("Mochila cheia! Descarte ou venda itens antes de comprar.")
        
    if jogador["moedas"] < preco:
        raise SaldoInsuficienteError(jogador["moedas"], preco, nome_item)
        
    jogador["moedas"] -= preco
    jogador["inventario"].append(nome_item)
    return f"Item '{nome_item}' adquirido com sucesso! Saldo restante: {jogador['moedas']} moedas."

# Testando o lançamento e captura especializada
player = {"nome": "Aragorn", "moedas": 100, "inventario": ["Espada", "Poção"]}

try:
    print(comprar_item_rpg(player, "Armadura Lendária", preco=250, slots_ocupados=2))
except SaldoInsuficienteError as err:
    print(f"⚠️ AVISO DA LOJA: {err}")
    print(f"👉 Dica: Você precisa de mais {err.deficit} moedas para este item.")
except InventarioCheioError as err:
    print(f"🎒 AVISO DE ESPAÇO: {err}")
````

````python
# Célula de Prática: Exercício 2
from typing import List, Dict, Tuple

def processar_leituras_temperatura(leituras_brutas: List[str]) -> Tuple[List[float], Dict[str, int]]:
    """Processa e valida uma lista de strings de leituras térmicas, registrando estatísticas de falhas.

    Parâmetros:
        leituras_brutas (List[str]): Lista com leituras brutas em formato string.

    Retorno:
        Tuple[List[float], Dict[str, int]]: Lista de floats válidos e dicionário com contagem de erros.
    """
    pass

# Exemplo de teste:
# dados_estacao = ["22.5", "24.0", "", "ERR_TIMEOUT", "150.0", "-12.3", "   ", "abc", "-80.0"]
# validos, erros = processar_leituras_temperatura(dados_estacao)
# print("Válidos:", validos)
# print("Erros:", erros)
#
# Saída Esperada:
# Válidos: [22.5, 24.0, -12.3]
# Erros: {'VAZIO': 2, 'CORROMPIDO': 2, 'FORA_DA_FAIXA': 2}
````

````python
# Resposta Comentada: Exercício 2
from typing import List, Dict, Tuple

def processar_leituras_temperatura(leituras_brutas: List[str]) -> Tuple[List[float], Dict[str, int]]:
    validos = []
    erros = {"VAZIO": 0, "CORROMPIDO": 0, "FORA_DA_FAIXA": 0}
    
    for item in leituras_brutas:
        texto = item.strip()
        try:
            if not texto:
                raise ValueError("VAZIO")
            
            valor = float(texto)
            if valor < -50.0 or valor > 60.0:
                raise ValueError("FORA_DA_FAIXA")
                
            validos.append(valor)
        except ValueError as err:
            motivo = str(err)
            if motivo == "VAZIO":
                erros["VAZIO"] += 1
            elif motivo == "FORA_DA_FAIXA":
                erros["FORA_DA_FAIXA"] += 1
            else:
                erros["CORROMPIDO"] += 1
                
    return (validos, erros)

# Testando a solução
dados_estacao = ["22.5", "24.0", "", "ERR_TIMEOUT", "150.0", "-12.3", "   ", "abc", "-80.0"]
validos, erros = processar_leituras_temperatura(dados_estacao)
print("Válidos:", validos)
print("Erros:", erros)
````

````python
# Resposta Comentada: Exercício 3
def validar_politica_senha(usuario: str, senha: str) -> bool:
    if len(senha) < 8:
        raise ValueError("A senha deve ter no mínimo 8 caracteres.")
        
    tem_numero = any(c.isdigit() for c in senha)
    if not tem_numero:
        raise ValueError("A senha deve conter pelo menos um número.")
        
    tem_maiuscula = any(c.isupper() for c in senha)
    if not tem_maiuscula:
        raise ValueError("A senha deve conter pelo menos uma letra maiúscula.")
        
    if usuario.lower() in senha.lower():
        raise PermissionError("A senha não pode conter o nome do usuário.")
        
    return True

# Testando a solução
testes = [
    ("carlos", "Car12345"),
    ("carlos", "curta1A"),
    ("carlos", "semnumeroA"),
    ("carlos", "sem_maiuscula_123"),
    ("carlos", "carlosSeguro123")
]

for usr, pwd in testes:
    try:
        validar_politica_senha(usr, pwd)
        print(f"Senha '{pwd}' para '{usr}': VÁLIDA ✅")
    except (ValueError, PermissionError) as e:
        print(f"Senha '{pwd}' para '{usr}': RECUSADA ❌ -> {e}")
````

### 🛒 Exercício 4 (E-commerce): Processador de Pedidos com `else` e `finally` (Log de Auditoria)

**Contexto:**
Um microsserviço de compras precisa debitar itens do inventário de uma loja online. A operação precisa garantir que:
1. O item exista no dicionário de estoque (caso contrário, lança `KeyError`).
2. A quantidade solicitada seja maior que zero (caso contrário, lança `ValueError("QUANTIDADE_INVALIDA")`).
3. O estoque atual seja suficiente (caso contrário, lança `ValueError("ESTOQUE_INSUFICIENTE")`).
4. Caso a compra seja aprovada (cláusula `else`), o estoque seja devidamente debitado.
5. Em **todas as tentativas** (com sucesso ou falha), um log de auditoria seja gravado na lista `registro_auditoria` (cláusula `finally`).

**Objetivo:**
Crie a função `processar_compra_estoque(estoque: dict[str, int], item: str, quantidade: int, auditoria: list[str]) -> tuple[bool, str]` que implementa esse fluxo completo utilizando `try / except / else / finally`.

````python
# Célula de Prática: Exercício 4
from typing import Dict, List, Tuple

def processar_compra_estoque(
    estoque: Dict[str, int], 
    item: str, 
    quantidade: int, 
    auditoria: List[str]
) -> Tuple[bool, str]:
    """Processa a compra de um item, atualizando o estoque e gravando log incondicional via finally.

    Parâmetros:
        estoque (Dict[str, int]): Dicionário com itens e quantidades disponíveis.
        item (str): Nome do produto solicitado.
        quantidade (int): Quantidade desejada.
        auditoria (List[str]): Lista onde serão anexados os registros de auditoria.

    Retorno:
        Tuple[bool, str]: Status de sucesso (True/False) e mensagem explicativa.
    """
    pass

# Exemplo de teste:
# estoque_loja = {"Notebook": 5, "Mouse": 20, "Teclado": 8}
# logs = []
#
# print(processar_compra_estoque(estoque_loja, "Notebook", 2, logs))
# print(processar_compra_estoque(estoque_loja, "Monitor", 1, logs))
# print(processar_compra_estoque(estoque_loja, "Mouse", 30, logs))
# print("\nHistórico de Auditoria:")
# for log in logs:
#     print("-", log)
#
# Saída Esperada:
# (True, 'COMPRA_APROVADA')
# (False, 'ITEM_NAO_ENCONTRADO')
# (False, 'ESTOQUE_INSUFICIENTE')
# Histórico de Auditoria:
# - Pedido Notebook (qtd: 2): SUCESSO | Estoque restante: 3
# - Pedido Monitor (qtd: 1): ERRO (ITEM_NAO_ENCONTRADO)
# - Pedido Mouse (qtd: 30): ERRO (ESTOQUE_INSUFICIENTE)
````

````python
# Resposta Comentada: Exercício 4
from typing import Dict, List, Tuple

def processar_compra_estoque(
    estoque: Dict[str, int], 
    item: str, 
    quantidade: int, 
    auditoria: List[str]
) -> Tuple[bool, str]:
    sucesso = False
    mensagem = ""
    
    try:
        if item not in estoque:
            raise KeyError("ITEM_NAO_ENCONTRADO")
        if quantidade <= 0:
            raise ValueError("QUANTIDADE_INVALIDA")
        if estoque[item] < quantidade:
            raise ValueError("ESTOQUE_INSUFICIENTE")
    except KeyError:
        sucesso = False
        mensagem = "ITEM_NAO_ENCONTRADO"
    except ValueError as err:
        sucesso = False
        mensagem = str(err)
    else:
        estoque[item] -= quantidade
        sucesso = True
        mensagem = "COMPRA_APROVADA"
    finally:
        if sucesso:
            log_msg = f"Pedido {item} (qtd: {quantidade}): SUCESSO | Estoque restante: {estoque[item]}"
        else:
            log_msg = f"Pedido {item} (qtd: {quantidade}): ERRO ({mensagem})"
        auditoria.append(log_msg)
        
    return (sucesso, mensagem)

# Testando a solução
estoque_loja = {"Notebook": 5, "Mouse": 20, "Teclado": 8}
logs = []

print(processar_compra_estoque(estoque_loja, "Notebook", 2, logs))
print(processar_compra_estoque(estoque_loja, "Monitor", 1, logs))
print(processar_compra_estoque(estoque_loja, "Mouse", 30, logs))

print("\nHistórico de Auditoria:")
for log in logs:
    print("-", log)
````

### 🏦 Exercício 5 (Fintech / Exceções Customizadas): Sistema de Saque Bancário Digital

**Contexto:**
Um banco digital precisa de um módulo de saques e transferências com exceções de domínio ricas e informativas:
1. **`ContaBloqueadaError(Exception)`**: Disparada se a conta estiver com o status `bloqueada: True`.
2. **`LimiteExcedidoError(Exception)`**: Disparada se o valor do saque for maior que o limite diário permitido para a conta. Deve conter os atributos `valor_solicitado` e `limite_maximo`.
3. **`SaldoInsuficienteError(Exception)`**: Disparada se o saldo da conta for menor que o valor solicitado. Deve conter os atributos `saldo_disponivel`, `valor_solicitado` e `deficit`.

````python
# Célula de Prática: Exercício 5
from typing import Dict, Any, Tuple

# 1. Defina aqui as classes de exceção customizadas
class ContaBloqueadaError(Exception):
    pass

class LimiteExcedidoError(Exception):
    pass

class SaldoInsuficienteError(Exception):
    pass

# 2. Implemente a função de saque
def executar_saque_conta(conta: Dict[str, Any], valor: float) -> Tuple[float, str]:
    """Realiza um saque validando regras bancárias e tratando exceções de domínio especializadas.

    Parâmetros:
        conta (Dict[str, Any]): Dicionário com dados da conta (titular, saldo, limite_diario, bloqueada).
        valor (float): Quantia a ser sacada em reais.

    Retorno:
        Tuple[float, str]: Saldo final da conta e mensagem descritiva do resultado.
    """
    pass

# Exemplo de teste:
# c1 = {"titular": "Marina", "saldo": 1000.0, "limite_diario": 500.0, "bloqueada": False}
# print(executar_saque_conta(c1, 300.0))   # (700.0, 'SAQUE_REALIZADO')
# print(executar_saque_conta(c1, 600.0))   # (700.0, 'ERRO_LIMITE: Solicitado R$ 600.00 > Limite R$ 500.00')
# print(executar_saque_conta(c1, 800.0))   # (700.0, 'ERRO_SALDO: Saldo R$ 700.00 insuficiente. Faltam R$ 100.00')
#
# c2 = {"titular": "Bruno", "saldo": 500.0, "limite_diario": 500.0, "bloqueada": True}
# print(executar_saque_conta(c2, 100.0))   # (500.0, 'ERRO_CONTA_BLOQUEADA: Operacao nao permitida')
````

````python
# Resposta Comentada: Exercício 5
from typing import Dict, Any, Tuple

# 1. Definição das classes de exceção customizadas
class ContaBloqueadaError(Exception):
    def __init__(self, titular: str):
        super().__init__(f"Conta de '{titular}' está bloqueada para movimentações.")

class LimiteExcedidoError(Exception):
    def __init__(self, valor_solicitado: float, limite_maximo: float):
        self.valor_solicitado = valor_solicitado
        self.limite_maximo = limite_maximo
        super().__init__(f"Valor solicitado R$ {valor_solicitado:.2f} excede o limite diário de R$ {limite_maximo:.2f}.")

class SaldoInsuficienteError(Exception):
    def __init__(self, saldo_disponivel: float, valor_solicitado: float):
        self.saldo_disponivel = saldo_disponivel
        self.valor_solicitado = valor_solicitado
        self.deficit = valor_solicitado - saldo_disponivel
        super().__init__(f"Saldo de R$ {saldo_disponivel:.2f} insuficiente. Faltam R$ {self.deficit:.2f}.")

# 2. Função de saque com tratamento
def executar_saque_conta(conta: Dict[str, Any], valor: float) -> Tuple[float, str]:
    try:
        if conta.get("bloqueada", False):
            raise ContaBloqueadaError(conta.get("titular", "Desconhecido"))
            
        if valor > conta.get("limite_diario", 0.0):
            raise LimiteExcedidoError(valor, conta.get("limite_diario", 0.0))
            
        if valor > conta.get("saldo", 0.0):
            raise SaldoInsuficienteError(conta.get("saldo", 0.0), valor)
            
        conta["saldo"] -= valor
        return (round(conta["saldo"], 2), "SAQUE_REALIZADO")
        
    except ContaBloqueadaError:
        return (conta["saldo"], "ERRO_CONTA_BLOQUEADA: Operacao nao permitida")
    except LimiteExcedidoError as e:
        return (conta["saldo"], f"ERRO_LIMITE: Solicitado R$ {e.valor_solicitado:.2f} > Limite R$ {e.limite_maximo:.2f}")
    except SaldoInsuficienteError as e:
        return (conta["saldo"], f"ERRO_SALDO: Saldo R$ {e.saldo_disponivel:.2f} insuficiente. Faltam R$ {e.deficit:.2f}")

# Testando a solução
c1 = {"titular": "Marina", "saldo": 1000.0, "limite_diario": 500.0, "bloqueada": False}
print(executar_saque_conta(c1, 300.0))
print(executar_saque_conta(c1, 600.0))
print(executar_saque_conta(c1, 800.0))

c2 = {"titular": "Bruno", "saldo": 500.0, "limite_diario": 500.0, "bloqueada": True}
print(executar_saque_conta(c2, 100.0))
````

### 📊 Exercício 6 (Engenharia de Dados / Pipeline Batch): Importador Resiliente de Arquivos

**Contexto:**
Um pipeline de Big Data recebe uma lista de registros brutos de clientes em formato de dicionários. O pipeline não pode parar quando encontra um registro defeituoso; em vez disso, ele deve processar todos os registros válidos, contabilizar as anomalias por categoria e gerar um relatório estatístico de qualidade dos dados.

**Campos Esperados de Cada Registro:**

* `"email"` (string contendo `'@'` e `'.'`).
* `"idade"` (inteiro entre 0 e 120).
* `"score"` (float entre 0.0 e 1000.0).

**Possíveis Falhas a Tratar:**
* Se faltar algum campo: captura `KeyError` e incrementa `"CAMPO_AUSENTE"`.
* Se o tipo for incompatível (ex: idade passada como lista): captura `TypeError` e incrementa `"TIPO_INVALIDO"`.
* Se o valor for semanticamente inválido (id $\le 0$, idade fora da faixa, email sem arroba/ponto, score fora de $[0, 1000]$): lança e captura `ValueError` e incrementa `"VALOR_INVALIDO"`.

**Objetivo:**
Crie a função `pipeline_limpeza_dados(lote_dados: list[dict]) -> dict[str, Any]` que processe a lista e retorne um dicionário consolidado com:
* `"total_recebidos"` (int): Quantidade total de registros no lote.
* `"validos"` (list[dict]): Lista com os registros perfeitamente válidos e padronizados.
* `"erros"` (dict[str, int]): Dicionário com contagem das chaves `"CAMPO_AUSENTE"`, `"TIPO_INVALIDO"`, `"VALOR_INVALIDO"`.
* `"taxa_qualidade_pct"` (float): Porcentagem de registros válidos sobre o total recebido, arredondada em duas casas decimais (caso a lista seja vazia, retorna `0.0`).

---


````python
# Célula de Prática: Exercício 6
from typing import List, Dict, Any

def pipeline_limpeza_dados(lote_dados: List[Dict[str, Any]]) -> Dict[str, Any]:
    """Processa um lote de registros com tolerância a falhas, categorizando erros e calculando SLA de qualidade.

    Parâmetros:
        lote_dados (List[Dict[str, Any]]): Lista de dicionários de clientes brutos.

    Retorno:
        Dict[str, Any]: Relatório estruturado com válidos, contagem de erros e taxa percentual de qualidade.
    """
    pass

# Exemplo de teste:
# lote_exemplo = [
#     {"id": 1, "email": "ana@empresa.com", "idade": 28, "score": 750.0},
#     {"id": 2, "email": "sem_arroba.com", "idade": 34, "score": 600.0},
#     {"id": 3, "idade": 45, "score": 800.0},  # Falta 'email'
#     {"id": 4, "email": "pedro@site.org", "idade": -5, "score": 500.0}, # Idade negativa
#     {"id": 5, "email": "lucas@web.com", "idade": "vinte", "score": 900.0}, # Idade tipo errado
#     {"id": 6, "email": "clara@gov.br", "idade": 31, "score": 820.5}
# ]
#
# relatorio = pipeline_limpeza_dados(lote_exemplo)
# print("Total:", relatorio["total_recebidos"])
# print("Válidos encontrados:", len(relatorio["validos"]))
# print("Erros por categoria:", relatorio["erros"])
# print("Taxa de Qualidade:", relatorio["taxa_qualidade_pct"], "%")
#
# Saída Esperada:
# Total: 6
# Válidos encontrados: 2
# Erros por categoria: {'CAMPO_AUSENTE': 1, 'TIPO_INVALIDO': 1, 'VALOR_INVALIDO': 2}
# Taxa de Qualidade: 33.33 %
````

````python
# Resposta Comentada: Exercício 6
from typing import List, Dict, Any

def pipeline_limpeza_dados(lote_dados: List[Dict[str, Any]]) -> Dict[str, Any]:
    total = len(lote_dados)
    if total == 0:
        return {
            "total_recebidos": 0,
            "validos": [],
            "erros": {"CAMPO_AUSENTE": 0, "TIPO_INVALIDO": 0, "VALOR_INVALIDO": 0},
            "taxa_qualidade_pct": 0.0
        }
        
    validos = []
    erros = {"CAMPO_AUSENTE": 0, "TIPO_INVALIDO": 0, "VALOR_INVALIDO": 0}
    
    for registro in lote_dados:
        try:
            # 1. Checagem de presença de campos obrigatórios
            if "id" not in registro or "email" not in registro or "idade" not in registro or "score" not in registro:
                raise KeyError("CAMPO_AUSENTE")
                
            reg_id = registro["id"]
            email = registro["email"]
            idade = registro["idade"]
            score = registro["score"]
            
            # 2. Checagem estrita de tipos
            if not isinstance(reg_id, int) or not isinstance(email, str) or not isinstance(idade, int) or not isinstance(score, (int, float)):
                raise TypeError("TIPO_INVALIDO")
                
            # 3. Checagem semântica de valores
            if reg_id <= 0:
                raise ValueError("VALOR_INVALIDO")
            if "@" not in email or "." not in email:
                raise ValueError("VALOR_INVALIDO")
            if idade < 0 or idade > 120:
                raise ValueError("VALOR_INVALIDO")
            if score < 0.0 or score > 1000.0:
                raise ValueError("VALOR_INVALIDO")
                
            # Registro totalmente validado
            validos.append({
                "id": reg_id,
                "email": email.strip().lower(),
                "idade": idade,
                "score": float(score)
            })
            
        except KeyError:
            erros["CAMPO_AUSENTE"] += 1
        except TypeError:
            erros["TIPO_INVALIDO"] += 1
        except ValueError:
            erros["VALOR_INVALIDO"] += 1
            
    taxa_qualidade = round((len(validos) / total) * 100, 2)
    
    return {
        "total_recebidos": total,
        "validos": validos,
        "erros": erros,
        "taxa_qualidade_pct": taxa_qualidade
    }

# Testando a solução
lote_exemplo = [
    {"id": 1, "email": "ana@empresa.com", "idade": 28, "score": 750.0},
    {"id": 2, "email": "sem_arroba.com", "idade": 34, "score": 600.0},
    {"id": 3, "idade": 45, "score": 800.0},
    {"id": 4, "email": "pedro@site.org", "idade": -5, "score": 500.0},
    {"id": 5, "email": "lucas@web.com", "idade": "vinte", "score": 900.0},
    {"id": 6, "email": "clara@gov.br", "idade": 31, "score": 820.5}
]

relatorio = pipeline_limpeza_dados(lote_exemplo)
print("Total:", relatorio["total_recebidos"])
print("Válidos encontrados:", len(relatorio["validos"]))
print("Erros por categoria:", relatorio["erros"])
print("Taxa de Qualidade:", relatorio["taxa_qualidade_pct"], "%")
````
