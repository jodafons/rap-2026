## 1. O que é o NumPy e o Objeto `ndarray`?

O **NumPy** é uma biblioteca de código aberto escrita em **C** e **Fortran** com interface em Python. Seu núcleo gira em torno do objeto **`ndarray`** (*N-dimensional array*).

### ⚡ Por que o NumPy é tão rápido comparado às listas do Python?

| Característica | Listas Nativas do Python (`list`) | Arrays do NumPy (`ndarray`) |
| :--- | :--- | :--- |
| **Tipagem dos Dados** | **Heterogênea**: Pode conter `int`, `str`, `float` misturados. O Python precisa checar o tipo de cada item em tempo de execução (*Type Overhead*). | **Homogênea**: Todos os elementos possuem estritamente o **mesmo tipo** de dado (ex: todos `float64` ou todos `int32`). |
| **Alocação na Memória** | **Dispersa**: A lista guarda ponteiros para objetos espalhados na memória RAM. | **Contígua**: Todos os bytes são armazenados em um único bloco contínuo de memória (aproveitamento máximo do cache da CPU). |
| **Operações Matemáticas** | Exige laços `for` lentos interpretados em Python. | Executa operações compiladas em C em blocos vetorizados (*SIMD - Single Instruction, Multiple Data*). |

```
Lista Python:   [ ptr1 ] ---> (Objeto Int 10)
                [ ptr2 ] ---> (Objeto Str 'Oi')
                [ ptr3 ] ---> (Objeto Float 3.14)

NumPy ndarray:  | 10 | 20 | 30 | 40 | 50 |  <--- Bloco contínuo e ultrarrápido em C!
```

## 4. Indexação, Fatiamento (*Slicing*) e Máscaras Booleanas

A manipulação de elementos no NumPy é extremamente flexível e poderosa.

### ✂️ Fatiamento 1D e 2D:
* **Em 1D**: Funciona similar às listas: `array[inicio:fim:passo]`.
* **Em 2D**: Selecionamos linhas e colunas simultaneamente separadas por vírgula:
  `matriz[linhas, colunas]` (ex: `matriz[0:2, 1:3]` extrai as 2 primeiras linhas e as colunas 1 e 2).

### 🎭 Máscaras Booleanas (*Boolean Indexing*):
Podemos aplicar condições lógicas diretamente no array (ex: `arr > 10`), gerando uma matriz de `True`/`False`. Ao passar essa máscara dentro dos colchetes (`arr[arr > 10]`), o NumPy filtra e retorna apenas os valores que satisfazem a condição!

### 🔍 Função `np.where(condicao, valor_se_verdadeiro, valor_se_falso)`:
Funciona como um `if/else` vetorizado de alta performance.


````python
matriz = np.array([
    [10, 25, 30, 45],
    [50, 15, 70, 85],
    [90, 35, 20, 60]
])

print("Matriz Original:\n", matriz)

# 1. Fatiamento: Linhas de 0 a 2 (exclusivo) e colunas de 1 a 3 (exclusivo)
sub_matriz = matriz[0:2, 1:3]
print("\nSub-região [0:2, 1:3]:\n", sub_matriz)

# 2. Máscara Booleana: Selecionar apenas números maiores ou iguais a 50
filtro = matriz >= 50
print("\nMáscara Booleana (>= 50):\n", filtro)
print("Valores filtrados:", matriz[filtro])

# 3. np.where: Se for >= 50 marca 'ALTO', senão 'BAIXO'
classificacao = np.where(matriz >= 50, "ALTO", "BAIXO")
print("\nClassificação com np.where:\n", classificacao)
````

````python
# ==============================================================================
# EXERCÍCIO 2 - SOLUÇÃO COMENTADA
# ==============================================================================

def simular_batalha_rpg(num_rodadas: int, seed: int = 42) -> dict:
    # 1. Configura a semente aleatória para garantir resultados reprodutíveis
    np.random.seed(seed)
    
    # 2. Gera matriz (num_rodadas x 3) com valores de 1 a 6 (dados d6)
    dados = np.random.randint(1, 7, size=(num_rodadas, 3))
    
    # 3. Soma os 3 dados de cada rodada ao longo das colunas (axis=1)
    soma_base = np.sum(dados, axis=1)
    
    # 4. Aplica as regras de dano com np.where vetorizado
    danos = soma_base.copy().astype(int)
    
    # Aplica Crítico (+5) se soma >= 15
    mascara_critico = soma_base >= 15
    danos = np.where(mascara_critico, danos + 5, danos)
    
    # Aplica Falha (dano 0) se soma <= 5
    mascara_falha = soma_base <= 5
    danos = np.where(mascara_falha, 0, danos)
    
    return {
        "rolagens_dados": dados,
        "danos_finais": danos,
        "dano_total": int(np.sum(danos)),
        "qtd_criticos": int(np.sum(mascara_critico)),
        "qtd_falhas": int(np.sum(mascara_falha))
    }

# Execução e Teste:
resultado_rpg = simular_batalha_rpg(num_rodadas=10, seed=42)
print("⚔️ Simulação de Batalha RPG (10 Rodadas):")
print("   • Danos por rodada:   ", resultado_rpg["danos_finais"])
print(f"   • Dano Total Causado:  {resultado_rpg['dano_total']} pontos")
print(f"   • Acertos Críticos:    {resultado_rpg['qtd_criticos']}")
print(f"   • Falhas Críticas:     {resultado_rpg['qtd_falhas']}")
````

````python
# ==============================================================================
# EXERCÍCIO 3 - SOLUÇÃO COMENTADA
# ==============================================================================

def analisar_radar_grid(tabuleiro: np.ndarray, lin: int, col: int, raio: int = 1) -> dict:
    num_linhas, num_cols = tabuleiro.shape
    
    # 1. Delimita as coordenadas de corte respeitando os limites da matriz
    lin_ini = max(0, lin - raio)
    lin_fim = min(num_linhas, lin + raio + 1)
    col_ini = max(0, col - raio)
    col_fim = min(num_cols, col + raio + 1)
    
    # 2. Fatiamento 2D da sub-região
    submatriz = tabuleiro[lin_ini:lin_fim, col_ini:col_fim]
    
    # 3. Agregações e contagem com máscara booleana
    pontos = int(np.sum(submatriz))
    qtd_alvos = int(np.sum(submatriz > 0))
    maior_alvo = int(np.max(submatriz)) if submatriz.size > 0 else 0
    
    return {
        "sub_regiao": submatriz,
        "pontos_totais": pontos,
        "qtd_alvos": qtd_alvos,
        "maior_alvo": maior_alvo
    }

# Execução e Teste:
mapa = np.array([
    [0,  0,  0,  0,  0],
    [0, 10,  0, 20,  0],
    [0,  0, 50,  0,  0],
    [0, 30,  0,  0, 40],
    [0,  0,  0,  0,  0]
])

radar_info = analisar_radar_grid(mapa, lin=2, col=2, raio=1)
print("📡 Varredura do Radar 3x3 no Centro (2, 2):")
print(radar_info["sub_regiao"])
print(f"   • Pontos Totais na Área: {radar_info['pontos_totais']}")
print(f"   • Alvos Detectados:     {radar_info['qtd_alvos']}")
print(f"   • Alvo de Maior Valor:   {radar_info['maior_alvo']}")
````

## 5. Vetorização, *Broadcasting* e Geometria Vetorial

### 📡 O que é *Broadcasting* no NumPy?
O *Broadcasting* permite que o NumPy realize operações aritméticas entre arrays de formatos (*shapes*) diferentes sem fazer cópias desnecessárias de memória.

**Exemplo:** Subtrair as coordenadas de um jogador $(1 	imes 2)$ de uma matriz com vários inimigos $(N 	imes 2)$. O NumPy automaticamente "estende" as coordenadas do jogador para subtrair de cada linha de uma só vez!

```
Inimigos (3 x 2):          Jogador (1 x 2):
[[10.0, 20.0],      -       [2.0, 3.0]
 [15.0, 30.0],
 [5.0,  12.0]]

Resultado (Diferenças dx e dy vetorizadas):
[[ 8.0, 17.0],
 [13.0, 27.0],
 [ 3.0,  9.0]]
```

### 📐 Distância Euclidiana em 2D:
Para cada objeto $(x_i, y_i)$, a distância até o ponto $(x_0, y_0)$ é calculada por:
$$d_i = \sqrt{(x_i - x_0)^2 + (y_i - y_0)^2}$$

````python
# ==============================================================================
# EXERCÍCIO 4 - SOLUÇÃO COMENTADA
# ==============================================================================

def detectar_proximidade_jogadores(
    posicao_jogador: np.ndarray, 
    posicoes_inimigos: np.ndarray, 
    raio_alerta: float = 5.0
) -> dict:
    # 1. Diferença de coordenadas usando broadcasting (N x 2 - 1 x 2)
    delta = posicoes_inimigos - posicao_jogador
    
    # 2. Distância euclidiana: sqrt(dx^2 + dy^2) somando ao longo do axis=1
    distancias = np.sqrt(np.sum(delta ** 2, axis=1))
    
    # 3. Encontra o índice de menor distância usando np.argmin()
    idx_mais_proximo = int(np.argmin(distancias))
    menor_dist = float(distancias[idx_mais_proximo])
    
    # 4. Filtra os inimigos dentro do raio usando máscara booleana
    mascara_alerta = distancias <= raio_alerta
    inimigos_perto = posicoes_inimigos[mascara_alerta]
    
    return {
        "distancias": distancias,
        "indice_mais_proximo": idx_mais_proximo,
        "menor_distancia": round(menor_dist, 2),
        "inimigos_no_raio": inimigos_perto
    }

# Execução e Teste:
player = np.array([10.0, 10.0])
inimigos = np.array([
    [12.0, 10.0],
    [20.0, 30.0],
    [13.0, 14.0],
    [10.0, 18.0]
])

prox_info = detectar_proximidade_jogadores(player, inimigos, raio_alerta=5.0)
print("🎮 Radar de Proximidade do Jogo:")
print("   • Distâncias calculadas:     ", np.round(prox_info["distancias"], 2))
print(f"   • Alvo Mais Próximo:          Inimigo #{prox_info['indice_mais_proximo']} a {prox_info['menor_distancia']}m")
print(f"   • Alvos em Raio de Ataque (<= 5m):\n{prox_info['inimigos_no_raio']}")
````

## 6. Agregações e Estatística por Eixos (`axis`)

Uma das operações mais frequentes em análise de dados é sumarizar matrizes em direções específicas usando o parâmetro **`axis`**:

* **Sem especificar `axis`**: Reduz a matriz inteira a um único número escalar.
* **`axis=0` (Ao longo das Linhas / Por Colunas)**: O cálculo "escorrega" verticalmente por cada coluna. O resultado tem tamanho igual ao número de colunas.
* **`axis=1` (Ao longo das Colunas / Por Linhas)**: O cálculo "escorrega" horizontalmente por cada linha. O resultado tem tamanho igual ao número de linhas.

```
                  axis = 0 (por coluna) ↓
                  Col 0    Col 1    Col 2
Linha 0  --->   [  10,      20,      30  ]   ---> axis = 1 (por linha)
Linha 1  --->   [  40,      50,      60  ]   ---> axis = 1 (por linha)
```

````python
# ==============================================================================
# EXERCÍCIO 5 - SOLUÇÃO COMENTADA
# ==============================================================================

def analisar_vendas_semana(matriz_vendas: np.ndarray) -> dict:
    # 1. Total faturado por filial (ao longo das colunas -> axis=1)
    total_filiais = np.sum(matriz_vendas, axis=1)
    
    # 2. Média de vendas da rede por dia da semana (ao longo das linhas -> axis=0)
    media_dias = np.mean(matriz_vendas, axis=0)
    
    # 3. Total diário para encontrar o dia mais lucrativo
    total_dias = np.sum(matriz_vendas, axis=0)
    dia_lucrativo = int(np.argmax(total_dias))
    
    # 4. Filial campeã de vendas (índice da filial de maior faturamento)
    filial_top = int(np.argmax(total_filiais))
    
    return {
        "total_por_filial": total_filiais,
        "media_por_dia": media_dias,
        "dia_mais_lucrativo": dia_lucrativo,
        "filial_campea": filial_top
    }

# Execução e Teste:
vendas = np.array([
    [1200, 1500, 1100, 1800, 2200, 3000, 2500],
    [ 900,  950, 1200, 1300, 1600, 2100, 1900],
    [2000, 2200, 1900, 2500, 3100, 4200, 3800],
    [1500, 1400, 1600, 1700, 2000, 2800, 2400]
])

dias_semana = ["Segunda", "Terça", "Quarta", "Quinta", "Sexta", "Sábado", "Domingo"]
lojas = ["Loja A", "Loja B", "Loja C (Top)", "Loja D"]

relat = analisar_vendas_semana(vendas)
print("📊 Relatório Semanal de Vendas da Rede:")
for i, tot in enumerate(relat["total_por_filial"]):
    print(f"   • {lojas[i]}: R$ {tot:,.2f}")
print(f"   • Dia mais lucrativo: {dias_semana[relat['dia_mais_lucrativo']]}")
print(f"   • Loja Campeã da Semana: {lojas[relat['filial_campea']]}")
````

## 7. Álgebra Linear e Resolução de Sistemas com `np.linalg`

O módulo **`np.linalg`** contém ferramentas matemáticas de alta precisão para resolver problemas lineares:

* **Multiplicação de Matrizes**: Operador `@` ou `np.matmul(A, B)`.
* **Transposição**: `A.T`.
* **Determinante**: `np.linalg.det(A)`.
* **Inversa**: `np.linalg.inv(A)`.
* **Resolução de Sistemas Lineares ($A \cdot x = b$)**: `np.linalg.solve(A, b)`.

---

### 🧮 Exemplo Matemático de Sistema Linear:
Imagine um sistema com 3 incógnitas ($M = 	ext{Maçã}, B = 	ext{Banana}, L = 	ext{Laranja}$):
$$\begin{cases}
3M + 2B + 1L = 20 \\
1M + 4B + 2L = 20 \\
2M + 1B + 3L = 17
\end{cases}$$

Em formato matricial:
$$\begin{bmatrix} 3 & 2 & 1 \\ 1 & 4 & 2 \\ 2 & 1 & 3 \end{bmatrix} \begin{bmatrix} M \\ B \\ L \end{bmatrix} = \begin{bmatrix} 20 \\ 20 \\ 17 \end{bmatrix}$$

Com o NumPy, resolvemos isso em uma única linha: `x = np.linalg.solve(A, b)`!

````python
# ==============================================================================
# EXERCÍCIO 6 - SOLUÇÃO COMENTADA
# ==============================================================================

def calcular_precos_unitarios(matriz_quantidades: np.ndarray, totais_pagos: np.ndarray) -> dict:
    # 1. Resolve o sistema linear A * x = b
    solucao = np.linalg.solve(matriz_quantidades, totais_pagos)
    
    # 2. Desempacota as incógnitas M, B, L
    preco_m = float(solucao[0])
    preco_b = float(solucao[1])
    preco_l = float(solucao[2])
    
    return {
        "preco_maca": round(preco_m, 2),
        "preco_banana": round(preco_b, 2),
        "preco_laranja": round(preco_l, 2)
    }

# Execução e Teste:
A = np.array([
    [3, 2, 1],
    [1, 4, 2],
    [2, 1, 3]
])
b = np.array([20.0, 20.0, 17.0])

tabela_precos = calcular_precos_unitarios(A, b)
print("🍎 Preços Unitários Descobertos via Álgebra Linear:")
print(f"   • Maçã:   R$ {tabela_precos['preco_maca']:.2f}")
print(f"   • Banana: R$ {tabela_precos['preco_banana']:.2f}")
print(f"   • Laranja: R$ {tabela_precos['preco_laranja']:.2f}")
````
