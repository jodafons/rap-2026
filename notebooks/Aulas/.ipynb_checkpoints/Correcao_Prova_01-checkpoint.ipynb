````python
# ==============================================================================
# QUESTÃO 9 - GABARITO OFICIAL
# ==============================================================================
from typing import List

def eh_primo_gabarito(numero: int) -> bool:
    # Casos base: números menores ou iguais a 1 não são primos
    if numero <= 1:
        return False
    
    # Testa possíveis divisores de 2 até a raiz quadrada de numero
    for divisor in range(2, int(numero**0.5) + 1):
        if numero % divisor == 0:
            return False  # Encontrou divisor: não é primo
            
    return True  # Não possui nenhum divisor além de 1 e ele mesmo


def filtrar_primos_intervalo_gabarito(inicio: int, fim: int) -> List[int]:
    primos = []
    
    # Itera no intervalo fechado [inicio, fim] (fim + 1 para incluir fim)
    for num in range(inicio, fim + 1):
        if eh_primo_gabarito(num):
            primos.append(num)
            
    return primos


# ------------------------------------------------------------------------------
# Validação Automatizada do Gabarito:
# ------------------------------------------------------------------------------
assert eh_primo_gabarito(0) == False
assert eh_primo_gabarito(1) == False
assert eh_primo_gabarito(2) == True
assert eh_primo_gabarito(3) == True
assert eh_primo_gabarito(4) == False
assert eh_primo_gabarito(9) == False
assert eh_primo_gabarito(25) == False
assert eh_primo_gabarito(29) == True

assert filtrar_primos_intervalo_gabarito(10, 30) == [11, 13, 17, 19, 23, 29]
assert filtrar_primos_intervalo_gabarito(1, 10) == [2, 3, 5, 7]
assert filtrar_primos_intervalo_gabarito(14, 16) == []
assert filtrar_primos_intervalo_gabarito(20, 20) == []
assert filtrar_primos_intervalo_gabarito(23, 23) == [23]

print("✅ [Gabarito Q9] Todos os testes passaram com sucesso!")
````

````python
# ==============================================================================
# QUESTÃO 10 - GABARITO OFICIAL
# ==============================================================================
from typing import List

def calcular_media_movel_gabarito(
    precos_diarios: List[float], 
    tamanho_janela: int
) -> List[float]:
    # 1. Validação de tamanho mínimo e janela válida
    if len(precos_diarios) < tamanho_janela or tamanho_janela <= 0:
        return []
        
    medias = []
    # Quantidade de janelas deslizantes completas: N - K + 1
    n_janelas = len(precos_diarios) - tamanho_janela + 1
    
    # 2. Deslocamento da janela e cálculo da média aritmética
    for i in range(n_janelas):
        janela = precos_diarios[i : i + tamanho_janela]
        media_janela = round(sum(janela) / tamanho_janela, 2)
        medias.append(media_janela)
        
    return medias


# ------------------------------------------------------------------------------
# Validação Automatizada do Gabarito:
# ------------------------------------------------------------------------------
assert calcular_media_movel_gabarito([100.0, 102.0, 104.0, 106.0, 108.0], 3) == [102.0, 104.0, 106.0]
assert calcular_media_movel_gabarito([50.0, 55.0], 3) == []
assert calcular_media_movel_gabarito([10.0, 20.0, 30.0], 1) == [10.0, 20.0, 30.0]
assert calcular_media_movel_gabarito([], 3) == []
assert calcular_media_movel_gabarito([10.0, 20.0], 0) == []
assert calcular_media_movel_gabarito([10.0, 20.0], -1) == []
assert calcular_media_movel_gabarito([10.0, 20.0, 30.0, 40.0], 4) == [25.0]

print("✅ [Gabarito Q10] Todos os testes passaram com sucesso!")
````

````python
# ==============================================================================
# QUESTÃO 11 (BÔNUS) - GABARITO OFICIAL
# ==============================================================================
from typing import List, Dict, Any, Tuple

def simular_expedicao_calabouco_gabarito(
    salas: List[Dict[str, Any]]
) -> Tuple[str, int, int, int]:
    hp, mana, pocoes, ouro = 100, 50, 3, 0
    HP_MAX, MANA_MAX = 100, 50
    
    for sala in salas:
        tipo = sala.get("tipo")
        
        if tipo == "MONSTRO":
            dano_base = sala.get("dano_base", 0)
            
            # (a) Feitiço neutralizador (gasta 15 mana e não sofre dano)
            if mana >= 15:
                mana -= 15
            else:
                # (b) Sem mana: combate corporal e sofre dano
                hp -= dano_base
                
                # (c) Faixa crítica: consome poção se disponível
                if 0 < hp <= 30 and pocoes > 0:
                    hp = min(HP_MAX, hp + 40)
                    pocoes -= 1
                    
            # (d) Morte / Derrota imediata
            if hp <= 0:
                return ("DERROTA", 0, mana, ouro)
                
        elif tipo == "DESCANSO":
            # Recupera mana respeitando o teto de 50
            mana = min(MANA_MAX, mana + sala.get("recupera_mp", 0))
            
        elif tipo == "TESOURO":
            # Coleta o ouro
            ouro += sala.get("ouro", 0)
            
    # Ao atravessar todas as salas viva
    return ("VITORIA", hp, mana, ouro)


# ------------------------------------------------------------------------------
# Validação Automatizada do Gabarito:
# ------------------------------------------------------------------------------
# 1. Caso do enunciado (Vitória com cura crítica)
teste_enunciado = [
    {"tipo": "MONSTRO", "dano_base": 30},
    {"tipo": "TESOURO", "ouro": 100},
    {"tipo": "MONSTRO", "dano_base": 30},
    {"tipo": "MONSTRO", "dano_base": 30},
    {"tipo": "MONSTRO", "dano_base": 80},
    {"tipo": "DESCANSO", "recupera_mp": 25}
]
assert simular_expedicao_calabouco_gabarito(teste_enunciado) == ("VITORIA", 60, 30, 100)

# 2. Caso de derrota fatal imediata
teste_derrota = [
    {"tipo": "MONSTRO", "dano_base": 20}, # gasta 15 mana (mana: 35)
    {"tipo": "MONSTRO", "dano_base": 20}, # gasta 15 mana (mana: 20)
    {"tipo": "MONSTRO", "dano_base": 20}, # gasta 15 mana (mana: 5)
    {"tipo": "MONSTRO", "dano_base": 120}, # sem mana, sofre 120 dano -> hp vira <= 0 imediatamente
    {"tipo": "TESOURO", "ouro": 500} # não deve coletar pois já foi derrotada
]
assert simular_expedicao_calabouco_gabarito(teste_derrota) == ("DERROTA", 0, 5, 0)

# 3. Caso com recuperação total de mana (teto 50)
teste_teto_mana = [
    {"tipo": "DESCANSO", "recupera_mp": 40},
    {"tipo": "TESOURO", "ouro": 50}
]
assert simular_expedicao_calabouco_gabarito(teste_teto_mana) == ("VITORIA", 100, 50, 50)

print("✅ [Gabarito Q11] Todos os testes passaram com sucesso!")
````
