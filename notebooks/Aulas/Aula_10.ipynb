### ⚠️ Limitações da Abordagem com Funções:
1. **Separação entre Dados e Comportamentos**: Os dados estão soltos em um dicionário e as funções estão soltas no código. Se tivermos dezenas de módulos, fica difícil saber quais funções pertencem a qual estrutura de dados.
2. **Falta de Proteção e Validação**: Qualquer parte do código pode acidentalmente alterar chaves do dicionário com valores inválidos (ex: `sub_func["oxigenio"] = -999`) ou digitar o nome de uma chave errado (`sub_func["oxigeno"]`), gerando erros sutis ou exceções `KeyError`.
3. **Sintaxe Menos Natural**: É necessário passar a variável explicitamente para toda função (`submergir(sub_func, 50.0)`) em vez de dizer ao próprio submarino para realizar a ação.

### ⚖️ Comparativo Lado a Lado: Funções vs. Classes

| Aspecto | Abordagem com Funções (Procedural) | Abordagem com Classes (Orientação a Objetos) |
| :--- | :--- | :--- |
| **União de Dados e Ações** | Dados (dicionários/variáveis) e funções estão separados | Dados (**atributos**) e funções (**métodos**) ficam reunidos no mesmo molde (**Encapsulamento**) |
| **Sintaxe de Execução** | `submergir(sub_func, 50.0)` *(passa o dado para a função externa)* | `sub_obj.submergir(50.0)` *(o próprio objeto executa a sua ação)* |
| **Criação de Entidades** | Requer funções auxiliares como `criar_submarino()` | Usa o método construtor padronizado `__init__` da própria classe |
| **Integridade dos Dados** | Dicionários são abertos e fáceis de corromper por engano | A classe centraliza e protege as regras de negócio e validações |
| **Extensibilidade e Reuso** | Difícil de especializar (exige muitos `if/else` soltos) | Permite usar **Herança** para criar tipos especializados sem duplicar código |

## 2. O que é Programação Orientada a Objetos (POO)?

Na **Orientação a Objetos**, nós modelamos nosso software como coleções de unidades chamadas **Objetos**, que contêm dados e comportamentos associados. Pensamos em termos de "coisas" ou "entidades" do mundo real.

### Conceitos Chave:
* **Classe**: É o molde, planta ou modelo a partir do qual os objetos são criados. Por exemplo: a planta de engenharia de um submarino.
* **Objeto**: É a instância física e concreta da classe. Por exemplo: o submarino real navegando nos oceanos, construído de acordo com a planta.
* **Atributo**: São as características ou dados que o objeto armazena (suas variáveis internas). Ex: `profundidade`, `oxigenio`, `nome`.
* **Método**: São as ações ou comportamentos que o objeto pode realizar (suas funções internas). Ex: `submergir()`, `emergir()`, `ligar_sistemas()`.

## 3. Entendendo a Sintaxe: O Construtor `__init__` e o Parâmetro `self`

Usamos a palavra-chave `class` para declarar uma classe. Por convenção, o nome da classe inicia com letra maiúscula no padrão **PascalCase** (ex: `Submarino`, `ReatorNuclear`).

### O construtor `__init__`:
* É o método construtor especial. Ele roda automaticamente sempre que criamos um novo objeto da classe (`sub = Submarino(...)`).
* Serve para definir e inicializar os atributos do objeto.

### O parâmetro `self`:
* O `self` representa **a instância específica do objeto que está executando aquele código**.
* Quando chamamos `sub_alfa.submergir(50.0)`, o Python converte por baixo dos panos para `Submarino.submergir(sub_alfa, 50.0)`.
* O primeiro parâmetro de qualquer método de instância dentro da classe é sempre o `self`, permitindo que o objeto leia e modifique seus próprios atributos (`self.nome`, `self.profundidade`).

### Instanciando (Criando) e Utilizando Múltiplos Objetos

Para criar um objeto a partir da classe, chamamos o nome da classe como se fosse uma função, passando os argumentos necessários definidos no `__init__`. A partir daí, podemos acessar os atributos e chamar os métodos do objeto usando o operador de ponto (`.`).

Veja abaixo como cada objeto criado mantém seu próprio estado de forma totalmente independente:

```python
# Definição consolidada da classe Submarino para as próximas seções:
class Submarino:
    def __init__(self, nome: str, tripulacao_maxima: int):
        self.nome = nome
        self.tripulacao_maxima = tripulacao_maxima
        self.profundidade = 0.0
        self.oxigenio = 100.0
        self.ligado = False

    def ligar_sistemas(self):
        self.ligado = True
        print(f"⚙️ Sistemas do submarino {self.nome} foram ativados!")

    def submergir(self, metros: float):
        if not self.ligado:
            print(f"⚠️ Não é possível submergir! Os sistemas do {self.nome} estão desligados.")
            return
        self.profundidade += metros
        self.oxigenio -= metros * 0.1
        print(f"⬇️ {self.nome} submergiu {metros}m. Profundidade atual: {self.profundidade}m. Oxigênio: {self.oxigenio:.1f}%")

    def emergir(self):
        self.profundidade = 0.0
        self.oxigenio = 100.0
        print(f"⬆️ {self.nome} subiu à superfície. Níveis de Oxigênio normalizados (100%).")

# Criando duas instâncias (objetos) diferentes da classe Submarino
sub_alfa = Submarino(nome="CIAA-LPS-Alpha", tripulacao_maxima=45)
sub_beta = Submarino(nome="CIAA-LPS-Beta", tripulacao_maxima=60)

# Acessando atributos diretamente
print(f"Submarino A: {sub_alfa.nome} | Tripulação Máxima: {sub_alfa.tripulacao_maxima}")
print(f"Submarino B: {sub_beta.nome} | Tripulação Máxima: {sub_beta.tripulacao_maxima}")

print("\n--- Testando Ações no Alfa ---")
# Chamando os métodos do submarino Alfa
sub_alfa.submergir(50.0) # Vai avisar que está desligado
sub_alfa.ligar_sistemas()
sub_alfa.submergir(50.0)
sub_alfa.submergir(30.0)
sub_alfa.emergir()

print("\n--- Verificando estado do Beta ---")
# Note que o estado do Beta permaneceu intacto, pois cada objeto tem seu próprio conjunto de variáveis!
print(f"Profundidade do {sub_beta.nome}: {sub_beta.profundidade}m")
print(f"Oxigênio do {sub_beta.nome}: {sub_beta.oxigenio:.1f}%")
```

## 4. Herança (Inheritance)

A **Herança** é um mecanismo que nos permite criar uma nova classe baseando-se em uma classe existente.
* A classe original é chamada de **Classe Mãe**, **Classe Base** ou **Superclasse**.
* A nova classe é chamada de **Classe Filha**, **Classe Derivada** ou **Subclasse**.

A subclasse herda automaticamente todos os atributos e métodos da superclasse, podendo adicionar novos comportamentos exclusivos ou modificar os existentes.

```python
# Superclasse (Classe Mãe)
class Embarcacao:
    def __init__(self, nome: str, velocidade_maxima_nos: float):
        self.nome = nome
        self.velocidade_maxima_nos = velocidade_maxima_nos
        self.motores_ligados = False

    def ligar_motores(self):
        self.motores_ligados = True
        print(f"🚀 Motores de {self.nome} ligados na potência máxima!")

    def desligar_motores(self):
        self.motores_ligados = False
        print(f"⏹️ Motores de {self.nome} desligados.")
```

### Criando as Subclasses

Para herdar de uma classe, passamos a classe mãe como parâmetro de parênteses na definição da classe filha: `class ClasseFilha(ClasseMae):`.

Se a classe filha precisar de um construtor próprio (para inicializar novos atributos específicos), chamamos o construtor da classe mãe usando a função `super().__init__(...)`.

```python
# Subclasse 1: Submarino de Ataque
class SubmarinoAtaque(Embarcacao): # Herda de Embarcacao
    def __init__(self, nome: str, velocidade_maxima_nos: float, torpedos_iniciais: int):
        # Inicializa o construtor da classe mãe (Embarcacao)
        super().__init__(nome, velocidade_maxima_nos)
        # Adiciona um atributo específico da classe filha
        self.torpedos = torpedos_iniciais

    # Método específico desta subclasse
    def disparar_torpedo(self):
        if not self.motores_ligados:
            print(f"⚠️ Não é possível disparar! Os motores do {self.nome} estão desligados.")
            return
        
        if self.torpedos > 0:
            self.torpedos -= 1
            print(f"💥 TORPEDO LANÇADO de {self.nome}! Impacto confirmado. Restam {self.torpedos} torpedos.")
        else:
            print(f"⚠️ Alerta! {self.nome} está sem munição de torpedos!")

# Subclasse 2: Submarino de Pesquisa Científica
class SubmarinoPesquisa(Embarcacao): # Herda de Embarcacao
    def __init__(self, nome: str, velocidade_maxima_nos: float, capacidade_amostras: int):
        super().__init__(nome, velocidade_maxima_nos)
        self.capacidade_amostras = capacidade_amostras
        self.amostras_coletadas = []

    # Método específico desta subclasse
    def coletar_amostra_marinha(self, tipo_amostra: str):
        if len(self.amostras_coletadas) < self.capacidade_amostras:
            self.amostras_coletadas.append(tipo_amostra)
            print(f"🔬 Amostra de '{tipo_amostra}' coletada com sucesso pelo braço robótico de {self.nome}.")
        else:
            print(f"⚠️ Compartimento de amostras do {self.nome} está completamente lotado!")
```

```python
# Instanciando os submarinos especializados
sub_combate = SubmarinoAtaque(nome="CIAA-Tubarao", velocidade_maxima_nos=32.0, torpedos_iniciais=4)
sub_cientifico = SubmarinoPesquisa(nome="CIAA-JacquesCousteau", velocidade_maxima_nos=15.5, capacidade_amostras=2)

# Ambos herdaram e usam os métodos da classe mãe (Embarcacao)
sub_combate.ligar_motores()
sub_cientifico.ligar_motores()

print("\n--- Testando Comportamentos Específicos ---")
# Testando métodos exclusivos da classe de combate
sub_combate.disparar_torpedo()
sub_combate.disparar_torpedo()

# Testando métodos exclusivos da classe de pesquisa
sub_cientifico.coletar_amostra_marinha("Coral Abissal")
sub_cientifico.coletar_amostra_marinha("Água Termal")
sub_cientifico.coletar_amostra_marinha("Peixe Lanterna") # Vai avisar que está cheio

# Note que tentar disparar torpedos do científico ou coletar amostras no combate gerará um erro,
# pois esses métodos não pertencem às respectivas classes filhas!
# sub_cientifico.disparar_torpedo() # AttributeError!
```

## 5. Sobrescrita de Métodos (Method Overriding)

O que acontece se uma subclasse precisar fazer a mesma ação que a classe mãe, mas de uma maneira ligeiramente diferente? Nós podemos simplesmente redefinir o método na subclasse com o **mesmo nome** do método na superclasse. Isso é chamado de **Sobrescrita de Método**.

Quando o método é chamado no objeto da subclasse, a versão sobrescrita (da filha) é executada em vez da versão original da classe mãe.

```python
# Superclasse base com método genérico
class EmbarcacaoBase:
    def __init__(self, nome: str):
        self.nome = nome

    def emitir_sinal(self):
        print(f"📢 {self.nome} emite um sinal de rádio padrão.")

# Sobrescrevendo o método emitir_sinal
class SubmarinoMilitar(EmbarcacaoBase):
    def emitir_sinal(self):
        # Sobrescreve por completo a ação da classe mãe
        print(f"🤫 {self.nome} está em modo silencioso. Não emitimos ondas de rádio para evitar detecção!")

class SubmarinoResgate(EmbarcacaoBase):
    def emitir_sinal(self):
        # Podemos chamar a versão da classe mãe usando super() antes ou depois da nossa lógica
        super().emitir_sinal()
        print(f"📡 RÁDIO EXTRA: [SOS] {self.nome} enviando coordenadas de emergência para a base CIAA!")
```

```python
militar = SubmarinoMilitar("CIAA-Espectral")
resgate = SubmarinoResgate("CIAA-SalvaVidas")

print("--- Chamada Militar ---")
militar.emitir_sinal()

print("\n--- Chamada Resgate ---")
resgate.emitir_sinal()
```

## 6. Métodos Especiais (Dunder Methods)

Em Python, métodos que começam e terminam com dois sublinhados (ex: `__init__`, `__str__`) são chamados de **Dunder Methods** (Double Underline) ou métodos especiais.

Um dos mais comuns é o método `__str__`. Ele define como o objeto será representado como uma string (texto) quando fizermos um `print(objeto)` ou usarmos `str(objeto)`.

```python
class Sonar:
    def __init__(self, modelo: str, alcance_metros: float):
        self.modelo = modelo
        self.alcance_metros = alcance_metros

    # Sem o __str__, o print exibiria apenas o endereço de memória do objeto (<__main__.Sonar object at ...>)
    def __str__(self):
        return f"📡 Sonar {self.modelo} | Alcance: {self.alcance_metros}m"

sonar_sub = Sonar(modelo="LPS-Sonar-2026", alcance_metros=5000.0)

# Ao printar, o Python executa automaticamente o método __str__
print(sonar_sub)
```
