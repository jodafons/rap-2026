### 🔹 Abordagem 1: Usando Funções e Dicionários (Paradigma Procedural)

No paradigma procedural, os **dados** (armazenados em dicionários) e as **funções** (que operam sobre esses dados) ficam totalmente separados.

### Exemplo 1 — Item a: Criando o Estado do Submarino

* **Nome da Função**: `criar_submarino(nome: str, tripulacao_maxima: int) -> dict`
* **Funcionalidade**: Cria e retorna um dicionário contendo o estado inicial do submarino.
* **Como fazer**: Retornar um dicionário com as seguintes chaves:
  * `"nome"`: string recebida por parâmetro.
  * `"tripulacao_maxima"`: inteiro recebido por parâmetro.
  * `"profundidade"`: float inicializado em `0.0` (metros).
  * `"oxigenio"`: float inicializado em `100.0` (porcentagem).
  * `"ligado"`: booleano inicializado como `False`.

```python
def criar_submarino(nome: str, tripulacao_maxima: int) -> dict:

    data = {
        "nome" : nome,
        "tripulacao_maxima" : tripulacao_maxima,
        "profundidade" : 0.0,
        "oxigenio" : 100.0,
        "ligado" : False, # maquina ligada (True) ou desligada (False)
    }
    return data

    

# Teste:
sub1 = criar_submarino("Submarino_1", 45)
print("Submarino criado:", sub1)
```

### Exemplo 1 — Item b: Ativando os Sistemas

* **Nome da Função**: `ligar_sistemas(submarino: dict)`
* **Funcionalidade**: Ativa os sistemas de bordo e os motores do submarino.
* **Como fazer**: Receber o dicionário `submarino`, atualizar a chave `submarino["ligado"] = True` e exibir a mensagem:
  `f"⚙️ Sistemas do submarino {submarino['nome']} foram ativados!"`.

```python
def ligar_sistemas(submarino: dict):
    if submarino["ligado"]==False:
        submarino["ligado"] = True

# Teste:
ligar_sistemas(sub1)
print("Status ligado:", sub1)
```

### Exemplo 1 — Item c: Submergindo o Submarino

* **Nome da Função**: `submergir(submarino: dict, metros: float)`
* **Funcionalidade**: Faz o submarino descer na água, recalculando a profundidade e consumindo oxigênio.
* **Como fazer**: 
  1. Verificar se `submarino["ligado"]` é `False`. Se estiver desligado, exibir `f"⚠️ Não é possível submergir! Os sistemas do {submarino['nome']} estão desligados."` e interromper a execução com `return`.
  2. Caso ligado, somar `metros` à chave `submarino["profundidade"]` e subtrair `metros * 0.1` de `submarino["oxigenio"]`.
  3. Exibir: `f"⬇️ {submarino['nome']} submergiu {metros}m. Profundidade atual: {submarino['profundidade']}m. Oxigênio: {submarino['oxigenio']:.1f}%"`.

```python
def submergir(submarino: dict, metros: float) -> bool:

    status_da_maquina = submarino["ligado"]
    nome = submarino['nome']
    if status_da_maquina == False:
        print(f"⚠️ Não é possível submergir! Os sistemas do {nome} estão desligados.")
        return False

              
    # se a maquina estiver ligada...
    submarino["profundidade"] += metros
    submarino["oxigenio"] -= metros*0.1
    print(f"⬇️ {submarino['nome']} submergiu {metros}m. Profundidade atual: {submarino['profundidade']}m. Oxigênio: {submarino['oxigenio']:.1f}%")
    return True

# Teste:
submergir(sub1, 50.0)
```

### Exemplo 1 — Item d: Emergindo à Superfície

* **Nome da Função**: `emergir(submarino: dict)`
* **Funcionalidade**: Retorna o submarino à superfície (profundidade 0) e normaliza os níveis de oxigênio para 100%.
* **Como fazer**: Redefinir `submarino["profundidade"] = 0.0`, `submarino["oxigenio"] = 100.0` e exibir a mensagem:
  `f"⬆️ {submarino['nome']} subiu à superfície. Níveis de Oxigênio normalizados (100%)."`.

```python
def emergir(submarino: dict):

    submarino["profundidade"] = 0.0
    submarino["oxigenio"] = 100
    
    print(f"⬆️ {submarino['nome']} subiu à superfície. Níveis de Oxigênio normalizados (100%).")
    
# Teste:
emergir(sub1)
```

### Exemplo 1 — Item e: Simulação Operacional Integrada

* **Funcionalidade**: Realizar o teste de uma missão completa utilizando as funções criadas nos itens anteriores.
* **Como fazer**:
  1. Criar o submarino `sub_alfa = criar_submarino("CIAA-LPS-Alpha", 45)`.
  2. Tentar submergir `50.0` metros com os sistemas desligados (observar a mensagem de aviso).
  3. Chamar `ligar_sistemas(sub_alfa)`.
  4. Submergir `50.0` metros e depois mais `30.0` metros.
  5. Chamar `emergir(sub_alfa)`.

```python
# Teste integrado das funções:
sub_alfa = criar_submarino("CIAA-LPS-Alpha", 45)
submergir(sub_alfa, 50.0)
ligar_sistemas(sub_alfa)
submergir(sub_alfa, 50.0)
submergir(sub_alfa, 30.0)
emergir(sub_alfa)
```

### ⚠️ Limitações da Abordagem com Funções:
1. **Separação entre Dados e Comportamentos**: Os dados estão soltos em um dicionário e as funções estão soltas no código. Se tivermos dezenas de módulos, fica difícil saber quais funções pertencem a qual estrutura de dados.
2. **Falta de Proteção e Validação**: Qualquer parte do código pode acidentalmente alterar chaves do dicionário com valores inválidos (ex: `sub_func["oxigenio"] = -999`) ou digitar o nome de uma chave errado (`sub_func["oxigeno"]`), gerando erros sutis ou exceções `KeyError`.
3. **Sintaxe Menos Natural**: É necessário passar a variável explicitamente para toda função (`submergir(sub_func, 50.0)`) em vez de dizer ao próprio submarino para realizar a ação.

A classe funciona como uma "planta de engenharia", e cada objeto criado a partir dela carrega consigo suas próprias variáveis internas e suas próprias funções. Vamos construir a classe `Submarino` passo a passo no **Exemplo 2**.

### Exemplo 2 — Item b: Método de Instância `ligar_sistemas`

* **Nome do Método**: `ligar_sistemas(self)`
* **Funcionalidade**: Ativa os sistemas internos da própria instância do submarino.
* **Como fazer**: Dentro da classe `Submarino`, definir o método de instância `def ligar_sistemas(self):`, atualizar `self.ligado = True` e exibir:
  `f"⚙️ Sistemas do submarino {self.nome} foram ativados!"`.

```python
class Submarino:
    def __init__(self, nome: str, tripulacao_maxima: int):
        self.nome = nome
        self.tripulacao_maxima = tripulacao_maxima
        self.profundidade = 0.0
        self.oxigenio = 100.0
        self.ligado = False

    def ligar_sistemas(self):
        if self.ligado == False:
            print(f"⚙️ Sistemas do submarino {self.nome} foram ativados!")
            self.ligado = True
        

# Teste:
sub_obj = Submarino("CIAA-LPS-Alpha", 45)
sub_obj.ligar_sistemas()
print("Sistemas ligados?", sub_obj.ligado)
```

### Exemplo 2 — Item c: Método de Instância `submergir`

* **Nome do Método**: `submergir(self, metros: float)`
* **Funcionalidade**: Faz o submarino descer na água, validando se está ligado e recalculando profundidade e oxigênio.
* **Como fazer**: Dentro da classe `Submarino`, definir `def submergir(self, metros: float):`:
  1. Verificar `if not self.ligado:`, exibir `f"⚠️ Não é possível submergir! Os sistemas do {self.nome} estão desligados."` e fazer `return`.
  2. Caso ligado, atualizar `self.profundidade += metros` e `self.oxigenio -= metros * 0.1`.
  3. Exibir: `f"⬇️ {self.nome} submergiu {metros}m. Profundidade atual: {self.profundidade}m. Oxigênio: {self.oxigenio:.1f}%"`.

```python
class Submarino:
    def __init__(self, nome: str, tripulacao_maxima: int):
        self.nome = nome
        self.tripulacao_maxima = tripulacao_maxima
        self.profundidade = 0.0
        self.oxigenio = 100.0
        self.ligado = False

    def ligar_sistemas(self) -> None:
        if self.ligado == False:
            print(f"⚙️ Sistemas do submarino {self.nome} foram ativados!")
            self.ligado = True

    def submergir( self, metros : float ) -> bool:

        if self.ligado == False:
            print (f"⚠️ Não é possível submergir! Os sistemas do {self.nome} estão desligados.")
            return False
        
        self.profundidade += metros 
        self.oxigenio -= metros*0.1
        print(f"⬇️ {self.nome} submergiu {metros}m. Profundidade atual: {self.profundidade}m. Oxigênio: {self.oxigenio:.1f}%")
        return True
    
# Teste:
sub_obj = Submarino("CIAA-LPS-Alpha", 45)
sub_obj.ligar_sistemas()
sub_obj.submergir(50.0)

```

### ⚖️ Comparativo Lado a Lado: Funções vs. Classes

| Aspecto | Abordagem com Funções (Procedural) | Abordagem com Classes (Orientação a Objetos) |
| :--- | :--- | :--- |
| **União de Dados e Ações** | Dados (dicionários/variáveis) e funções estão separados | Dados (**atributos**) e funções (**métodos**) ficam reunidos no mesmo molde (**Encapsulamento**) |
| **Sintaxe de Execução** | `submergir(sub_func, 50.0)` *(passa o dado para a função externa)* | `sub_obj.submergir(50.0)` *(o próprio objeto executa a sua ação)* |
| **Criação de Entidades** | Requer funções auxiliares como `criar_submarino()` | Usa o método construtor padronizado `__init__` da própria classe |
| **Integridade dos Dados** | Dicionários são abertos e fáceis de corromper por engano | A classe centraliza e protege as regras de negócio e validações |
| **Extensibilidade e Reuso** | Difícil de especializar (exige muitos `if/else` soltos) | Permite usar **Herança** para criar tipos especializados sem duplicar código |

## 2. O que é Programação Orientada a Objetos (POO)?

Na **Orientação a Objetos**, nós modelamos nosso software como coleções de unidades chamadas **Objetos**, que contêm dados e comportamentos associados. Pensamos em termos de "coisas" ou "entidades" do mundo real.

### Conceitos Chave:
* **Classe**: É o molde, planta ou modelo a partir do qual os objetos são criados. Por exemplo: a planta de engenharia de um submarino.
* **Objeto**: É a instância física e concreta da classe. Por exemplo: o submarino real navegando nos oceanos, construído de acordo com a planta.
* **Atributo**: São as características ou dados que o objeto armazena (suas variáveis internas). Ex: `profundidade`, `oxigenio`, `nome`.
* **Método**: São as ações ou comportamentos que o objeto pode realizar (suas funções internas). Ex: `submergir()`, `emergir()`, `ligar_sistemas()`.

## 3. Entendendo a Sintaxe: O Construtor `__init__` e o Parâmetro `self`

Usamos a palavra-chave `class` para declarar uma classe. Por convenção, o nome da classe inicia com letra maiúscula no padrão **PascalCase** (ex: `Submarino`, `ReatorNuclear`).

### O construtor `__init__`:
* É o método construtor especial. Ele roda automaticamente sempre que criamos um novo objeto da classe (`sub = Submarino(...)`).
* Serve para definir e inicializar os atributos do objeto.

### O parâmetro `self`:
* O `self` representa **a instância específica do objeto que está executando aquele código**.
* Quando chamamos `sub_alfa.submergir(50.0)`, o Python converte por baixo dos panos para `Submarino.submergir(sub_alfa, 50.0)`.
* O primeiro parâmetro de qualquer método de instância dentro da classe é sempre o `self`, permitindo que o objeto leia e modifique seus próprios atributos (`self.nome`, `self.profundidade`).

### Instanciando (Criando) e Utilizando Múltiplos Objetos

Para criar um objeto a partir da classe, chamamos o nome da classe como se fosse uma função, passando os argumentos necessários definidos no `__init__`. A partir daí, podemos acessar os atributos e chamar os métodos do objeto usando o operador de ponto (`.`).

Veja abaixo como cada objeto criado mantém seu próprio estado de forma totalmente independente:

```python
# Definição consolidada da classe Submarino para as próximas seções:
class Submarino:
    def __init__(self, nome: str, tripulacao_maxima: int):
        self.nome = nome
        self.tripulacao_maxima = tripulacao_maxima
        self.profundidade = 0.0
        self.oxigenio = 100.0
        self.ligado = False

    def ligar_sistemas(self):
        self.ligado = True
        print(f"⚙️ Sistemas do submarino {self.nome} foram ativados!")

    def submergir(self, metros: float):
        if not self.ligado:
            print(f"⚠️ Não é possível submergir! Os sistemas do {self.nome} estão desligados.")
            return
        self.profundidade += metros
        self.oxigenio -= metros * 0.1
        print(f"⬇️ {self.nome} submergiu {metros}m. Profundidade atual: {self.profundidade}m. Oxigênio: {self.oxigenio:.1f}%")

    def emergir(self):
        self.profundidade = 0.0
        self.oxigenio = 100.0
        print(f"⬆️ {self.nome} subiu à superfície. Níveis de Oxigênio normalizados (100%).")

# Criando duas instâncias (objetos) diferentes da classe Submarino
sub_alfa = Submarino(nome="CIAA-LPS-Alpha", tripulacao_maxima=45)
sub_beta = Submarino(nome="CIAA-LPS-Beta", tripulacao_maxima=60)

# Acessando atributos diretamente
print(f"Submarino A: {sub_alfa.nome} | Tripulação Máxima: {sub_alfa.tripulacao_maxima}")
print(f"Submarino B: {sub_beta.nome} | Tripulação Máxima: {sub_beta.tripulacao_maxima}")

print("\n--- Testando Ações no Alfa ---")
# Chamando os métodos do submarino Alfa
sub_alfa.submergir(50.0) # Vai avisar que está desligado
sub_alfa.ligar_sistemas()
sub_alfa.submergir(50.0)
sub_alfa.submergir(30.0)
sub_alfa.emergir()

print("\n--- Verificando estado do Beta ---")
# Note que o estado do Beta permaneceu intacto, pois cada objeto tem seu próprio conjunto de variáveis!
print(f"Profundidade do {sub_beta.nome}: {sub_beta.profundidade}m")
print(f"Oxigênio do {sub_beta.nome}: {sub_beta.oxigenio:.1f}%")
```

## 4. Herança (Inheritance)

A **Herança** é um mecanismo que nos permite criar uma nova classe baseando-se em uma classe existente.
* A classe original é chamada de **Classe Mãe**, **Classe Base** ou **Superclasse**.
* A nova classe é chamada de **Classe Filha**, **Classe Derivada** ou **Subclasse**.

A subclasse herda automaticamente todos os atributos e métodos da superclasse, podendo adicionar novos comportamentos exclusivos ou modificar os existentes.

```python
# Superclasse (Classe Mãe)
class Embarcacao:
    def __init__(self, nome: str, velocidade_maxima_nos: float):
        self.nome = nome
        self.velocidade_maxima_nos = velocidade_maxima_nos
        self.motores_ligados = False

    def ligar_motores(self):
        self.motores_ligados = True
        print(f"🚀 Motores de {self.nome} ligados na potência máxima!")

    def desligar_motores(self):
        self.motores_ligados = False
        print(f"⏹️ Motores de {self.nome} desligados.")
```

### Criando as Subclasses

Para herdar de uma classe, passamos a classe mãe como parâmetro de parênteses na definição da classe filha: `class ClasseFilha(ClasseMae):`.

Se a classe filha precisar de um construtor próprio (para inicializar novos atributos específicos), chamamos o construtor da classe mãe usando a função `super().__init__(...)`.

```python
# Superclasse (Classe Mãe)
class Embarcacao:
    def __init__(self, nome: str, velocidade_maxima_nos: float):
        self.nome = nome
        self.velocidade_maxima_nos = velocidade_maxima_nos
        self.motores_ligados = False

    def ligar_motores(self):
        self.motores_ligados = True
        print(f"🚀 Motores de {self.nome} ligados na potência máxima!")

    def desligar_motores(self):
        self.motores_ligados = False
        print(f"⏹️ Motores de {self.nome} desligados.")


# Subclasse 1: Submarino de Ataque
class SubmarinoAtaque(Embarcacao): # Herda de Embarcacao
    
    
    def __init__(self, nome: str, velocidade_maxima_nos: float, torpedos_iniciais: int):
        # Inicializa o construtor da classe mãe (Embarcacao)

        # self.nome 
        #super().__init__(nome, velocidade_maxima_nos)
        Embarcacao.__init__(self, nome, velocidade_maxima_nos)

        # Adiciona um atributo específico da classe filha
        self.torpedos = torpedos_iniciais

    # Método específico desta subclasse
    def disparar_torpedo(self):
        if not self.motores_ligados:
            print(f"⚠️ Não é possível disparar! Os motores do {self.nome} estão desligados.")
            return
        
        if self.torpedos > 0:
            self.torpedos -= 1
            print(f"💥 TORPEDO LANÇADO de {self.nome}! Impacto confirmado. Restam {self.torpedos} torpedos.")
        else:
            print(f"⚠️ Alerta! {self.nome} está sem munição de torpedos!")

# Subclasse 2: Submarino de Pesquisa Científica
class SubmarinoPesquisa(Embarcacao): # Herda de Embarcacao
    def __init__(self, nome: str, velocidade_maxima_nos: float, capacidade_amostras: int):
        #super().__init__(nome, velocidade_maxima_nos)
           
        Embarcacao.__init__(self, nome, velocidade_maxima_nos)

        self.capacidade_amostras = capacidade_amostras
        self.amostras_coletadas = []

    # Método específico desta subclasse
    def coletar_amostra_marinha(self, tipo_amostra: str):
        if len(self.amostras_coletadas) < self.capacidade_amostras:
            self.amostras_coletadas.append(tipo_amostra)
            print(f"🔬 Amostra de '{tipo_amostra}' coletada com sucesso pelo braço robótico de {self.nome}.")
        else:
            print(f"⚠️ Compartimento de amostras do {self.nome} está completamente lotado!")
```

## 5. Sobrescrita de Métodos (Method Overriding)

O que acontece se uma subclasse precisar fazer a mesma ação que a classe mãe, mas de uma maneira ligeiramente diferente? Nós podemos simplesmente redefinir o método na subclasse com o **mesmo nome** do método na superclasse. Isso é chamado de **Sobrescrita de Método**.

Quando o método é chamado no objeto da subclasse, a versão sobrescrita (da filha) é executada em vez da versão original da classe mãe.

```python
# Superclasse base com método genérico
class EmbarcacaoBase:
    def __init__(self, nome: str):
        self.nome = nome

    def emitir_sinal(self):
        print(f"📢 {self.nome} emite um sinal de rádio padrão.")

# Sobrescrevendo o método emitir_sinal
class SubmarinoMilitar(EmbarcacaoBase):
    def emitir_sinal(self):
        # Sobrescreve por completo a ação da classe mãe
        print(f"🤫 {self.nome} está em modo silencioso. Não emitimos ondas de rádio para evitar detecção!")

class SubmarinoResgate(EmbarcacaoBase):
    def emitir_sinal(self):
        # Podemos chamar a versão da classe mãe usando super() antes ou depois da nossa lógica
        super().emitir_sinal()
        print(f"📡 RÁDIO EXTRA: [SOS] {self.nome} enviando coordenadas de emergência para a base CIAA!")
```

```python
militar = SubmarinoMilitar("CIAA-Espectral")
resgate = SubmarinoResgate("CIAA-SalvaVidas")

print("--- Chamada Militar ---")
militar.emitir_sinal()

print("\n--- Chamada Resgate ---")
resgate.emitir_sinal()
```

## 6. Métodos Especiais (Dunder Methods)

Em Python, métodos que começam e terminam com dois sublinhados (ex: `__init__`, `__str__`) são chamados de **Dunder Methods** (Double Underline) ou métodos especiais.

Um dos mais comuns é o método `__str__`. Ele define como o objeto será representado como uma string (texto) quando fizermos um `print(objeto)` ou usarmos `str(objeto)`.

```python
class Sonar:
    def __init__(self, modelo: str, alcance_metros: float):
        self.modelo = modelo
        self.alcance_metros = alcance_metros

    # Sem o __str__, o print exibiria apenas o endereço de memória do objeto (<__main__.Sonar object at ...>)
    def __str__(self):
        return f"📡 Sonar {self.modelo} | Alcance: {self.alcance_metros}m"

sonar_sub = Sonar(modelo="LPS-Sonar-2026", alcance_metros=5000.0)

# Ao printar, o Python executa automaticamente o método __str__
print(sonar_sub)
```

1. Crie uma classe `ReatorNuclear` com os atributos `codigo: str`, `temperatura: float` e `ativo: bool` (inicializado como `True`).
2. Crie um método `aumentar_temperatura(self, valor: float) -> None` que adiciona o valor à temperatura atual e imprime a nova temperatura formatada com 1 casa decimal.
3. Crie um método `resfriar(self, valor: float) -> None` que diminui o valor da temperatura atual e imprime a nova temperatura.
4. Crie um método `verificar_seguranca(self) -> bool` que verifica se a temperatura ultrapassou 300.0°C:
   * Se ultrapassar (> 300.0), imprime `"⚠️ ALERTA CRÍTICO: Reator [codigo] superaquecido ([temperatura]°C)! Reduzindo temperatura automaticamente..."`, chama `self.resfriar(50.0)`, define `self.ativo = False`, imprime `"🛑 Reator [codigo] desligado por protocolo de emergência."` e retorna `False`.
   * Caso contrário, imprime `"✅ Reator [codigo] operando dentro das normas de segurança ([temperatura]°C)."` e retorna `True`.

```python
# IMPLEMENTAÇÃO DO EXERCÍCIO 2

# Definição da classe ReatorNuclear
class ReatorNuclear:

    # construtor da classe ReatorNuclear
    def __init__(self, codigo : str,  temperatura : float, ativo : bool = True ):

        self.codigo = codigo
        self.temperatura =  temperatura
        self.ativo = ativo

    def aumentar_temperatura( self, valor : float ) -> None:
        self.temperatura += valor
        temperatura = round(self.temperatura, 1)
        print(f"Temperatura atual do reator é de {temperatura}" )

    def resfriar(self , valor : float) -> None:
        self.temperatura -= valor
        temperatura = round(self.temperatura, 1)
        print(f"Temperatura atual do reator é de {temperatura}" )

    def verificar_seguranca(self) -> bool:

        if self.temperatura > 300 :
            print(f"⚠️ ALERTA CRÍTICO: Reator {self.codigo} superaquecido ({self.temperatura}°C)! Reduzindo temperatura automaticamente...")
            self.resfriar(50.0)
            self.ativo = False
            print(f"🛑 Reator {self.codigo} desligado por protocolo de emergência.")
            return False 
        else:
            print(f"✅ Reator [codigo] operando dentro das normas de segurança ({self.temperatura}°C).")
            return True

 
```

### Exercício 3: Transmissor Tático e Criptografia de Mensagens (Herança + Manipulação de Strings + Listas)

**Cenário**: O submarino precisa enviar mensagens de rádio e mensagens acústicas táticas para a base naval, garantindo histórico e criptografia quando em águas hostis.

**Instruções**:
1. **Superclasse `TransmissorBase`**:
   * `__init__(self, identificador: str, frequencia_khz: float) -> None`: Inicializa `self.identificador: str`, `self.frequencia_khz: float` e `self.historico_mensagens: List[str] = []`.
   * `formatar_cabecalho(self) -> str`: Retorna a string de cabeçalho no formato `f"[{self.identificador.strip().upper()} | {self.frequencia_khz:.1f}kHz]"`.
   * `enviar_mensagem(self, texto: str) -> str`: Formata o texto em maiúsculas e sem espaços nas pontas (`texto.strip().upper()`), concatena com o cabeçalho (`f"{self.formatar_cabecalho()} {texto_limpo}"`), adiciona a mensagem completa em `self.historico_mensagens` e retorna a string gerada.
   * `buscar_mensagens_por_termo(self, termo: str) -> List[str]`: Retorna uma lista contendo apenas as mensagens do histórico que contêm o `termo` (busca case-insensitive).

2. **Subclasse `TransmissorCriptografado(TransmissorBase)`**:
   * Herda de `TransmissorBase`.
   * `__init__(self, identificador: str, frequencia_khz: float, chave_deslocamento: int) -> None`: Inicializa a classe mãe via `super().__init__(identificador, frequencia_khz)` e adiciona o atributo `self.chave_deslocamento: int = chave_deslocamento`.
   * Método privado/auxiliar `cifrar_caracteres(self, texto: str) -> str`: Percorre a string; para cada caractere alfabético (`char.isalpha()`), desloca o caractere em `self.chave_deslocamento` posições no alfabeto (ex: para letras maiúsculas: `chr((ord(char) - ord('A') + self.chave_deslocamento) % 26 + ord('A'))`), mantendo caracteres não alfabéticos inalterados.
   * Sobrescreva `enviar_mensagem(self, texto: str) -> str`: Limpa o texto (`texto.strip().upper()`), aplica `cifrar_caracteres`, monta a mensagem no formato `f"{self.formatar_cabecalho()} [SIGILOSO] {texto_cifrado}"`, adiciona ao histórico e retorna a mensagem.

* **Exigência**: Utilize anotações de tipo (**Type Hints**) completas em todos os parâmetros, atributos e retornos.

```python
# IMPLEMENTAÇÃO DO EXERCÍCIO 3
from typing import List

# 1. Superclasse TransmissorBase
class TransmissorBase:


    def __init__(self, identificador: str, frequencia_khz: float ):

        # propriedades da classe transmissor base
        self.identificador = identificador 
        self.frequencia_khz = frequencia_khz
        self.historico_mensagens = []

    def formatar_cabecalho(self) -> str:
        identificador = self.identificador.strip().upper()
        frequencia = round( self.frequencia_khz , 1 )
        return f"[{identificador} | {frequencia}kHz]"

    def enviar_mensagem(self, texto: str) -> str:

        mensagem = texto.strip().upper()
        #cabecalho = self.formatar_cabecalho()
        output = f"{self.formatar_cabecalho()} {mensagem}"
        self.historico_mensagens.append( output )
        return output

    def buscar_mensagens_por_termo(self, termo: str) -> List[str]:
        termo = termo.upper()
        buscar_termos = []
        for mensagem in self.historico_mensagens:
            if termo in mensagem:
                buscar_termos.append(mensagem)

        return buscar_termos
                
        
    

# 2. Subclasse TransmissorCriptografado


class TransmissorCriptografado(TransmissorBase):

    def __init__(self, identificador: str, frequencia_khz: float, chave_deslocamento: int): 

        TransmissorBase.__init__(self, identificador, frequencia_khz)
        
        self.chave_deslocamento = chave_deslocamento

    def enviar_mensagem(self, texto: str) -> str:

        texto_cifrado = self.cifrar_caracteres( texto )
        
        return f"{self.formatar_cabecalho()} [SIGILOSO] {texto_cifrado}"

    
    def cifrar_caracteres(self, texto: str) -> str:

        # ABCDE
        resultado = []
        for char in texto:
            if char.isalpha():
                base: int = ord('A') if char.isupper() else ord('a')
                char_cifrado: str = chr((ord(char) - base + self.chave_deslocamento) % 26 + base)
                resultado.append(char_cifrado)
            else:
                 resultado.append(char)
        #print(resultado)
        return "".join(resultado)
 
        




# Teste do Exercício 3:
radio = TransmissorBase("CIAA-Radio-1", 142.5)
print(radio.enviar_mensagem("contato sonar identificado"))
```

**Cenário**: O sistema de sonar passivo cataloga alvos acústicos em um dicionário estruturado e analisa suas frequências para determinar se são biológicos, navios mercantes ou submarinos de guerra.

**Instruções**:
1. **Superclasse `SensorAcustico`**:
   * `__init__(self, nome_sensor: str, alcance_km: float) -> None`: Inicializa `self.nome_sensor: str`, `self.alcance_km: float` e o dicionário `self.contatos_detectados: Dict[str, Dict[str, Union[str, float]]] = {}`.
   * `registrar_contato(self, contato_id: str, tipo: str, distancia_km: float) -> bool`:

     2. Se `distancia_km > self.alcance_km`, imprime `"⚠️ Alvo [contato_id] fora do alcance de [nome_sensor]!"` e retorna `False`.
     3. Armazena no dicionário `self.contatos_detectados[id_limpo] = {"tipo": tipo.strip().upper(), "distancia_km": distancia_km}` e retorna `True`.
   * `listar_contatos_por_tipo(self, tipo_filtro: str) -> List[str]`: Retorna uma lista de strings contendo os `contato_id` cujo `"tipo"` coincide com `tipo_filtro.strip().upper()`.

```python
# IMPLEMENTAÇÃO DO EXERCÍCIO 5
from typing import List, Dict, Union, Any

# 1. Superclasse SensorAcustico
# __init__(self, nome_sensor: str, alcance_km: float) -> None: Inicializa self.nome_sensor: str, self.alcance_km: float e o 
# dicionário self.contatos_detectados: Dict[str, Dict[str, Union[str, float]]] = {}.
class SensorAcustico:

    def __init__(self, nome_sensor : str, alcance_km : float ) -> None: 
        self.nome_sensor = nome_sensor # atributo de uma classe
        self.alcance_km = alcance_km # atributo de uma classe
        self.contatos_detectados = {}
        # self.contatos_detector : Dict[str, Dict[str, Union[str,float]]] = {}
    

    # registrar_contato(self, contato_id: str, tipo: str, distancia_km: float) -> bool:
    # Limpa o contato_id com contato_id.strip().upper().
    # Se distancia_km > self.alcance_km, imprime "⚠️ Alvo [contato_id] fora do alcance de [nome_sensor]!" e retorna False.
    # Armazena no dicionário self.contatos_detectados[id_limpo] = {"tipo": tipo.strip().upper(), 
    # "distancia_km": distancia_km} e retorna True.
    def registrar_contato(self , contato_id : str, tipo : str, distancia_km : float ) -> bool : # nome disso é protótipo da função/método

        if distancia_km > self.alcance_km:
            print(f"⚠️ Alvo [{contato_id}] fora do alcance de [{self.nome_sensor}]!")
            return False

        contato = {
            "tipo"         : tipo.strip().upper(), # " xyz\n " -> "xyz" -> "XYZ"
            "distancia_km" : distancia_km,
            # assinatura : BILOGICO
        }

        self.contatos_detectados[contato_id] = contato
        return True

    # listar_contatos_por_tipo(self, tipo_filtro: str) -> List[str]: Retorna uma lista de strings contendo 
    # os contato_id cujo "tipo" coincide com tipo_filtro.strip().upper().
    # " xyz "-> "XYZ"
    def listar_contatos_por_tipo (self, tipo_filtro : str= None) -> List[str]:

        
        lista_de_contato_ids = []

        for contato_id in self.contatos_detectados.keys():
            contato = self.contatos_detectados[contato_id]
            
            if tipo_filtro != None:
                if contato["tipo"] == tipo_filtro.strip().upper():
                    lista_de_contato_ids.append( contato_id )
            else:
                lista_de_contato_ids.append(contato_id) 
                
        return lista_de_contato_ids
        
        
    

    
# 2. Subclasse SonarPassivoAvancado
class  SonarPassivoAvancado(SensorAcustico):

    # __init__(self, nome_sensor: str, alcance_km: float, sensibilidade_db: float) -> None: 
    # Inicializa com super() e adiciona self.sensibilidade_db: float = sensibilidade_db e 
    # self.assinaturas_frequencia: Dict[str, str] = {}.
    def __init__(self, nome_sensor : str , alcance_km : float, sensibilidade_db : float ) -> None: 
        
        SensorAcustico.__init__(self,  nome_sensor, alcance_km )
        self.sensibilidade_db = sensibilidade_db 
        self.assinaturas_frequencia : Dict[str, str] = {}


    # Método analisar_frequencia(self, contato_id: str, frequencia_hz: float) -> str:
    # Se frequencia_hz < 120.0, classifica como "PROPULSAO_PESADA".
    # Se 120.0 <= frequencia_hz <= 800.0, classifica como "SUBMARINO_DIESEL".
    # Se frequencia_hz > 800.0, classifica como "BIOLOGICO".
    # Salva em self.assinaturas_frequencia[contato_id.strip().upper()] = classificacao e retorna a classificação.
    def analisar_frequencia( self, contato_id : str , frequencia_hz : float ) -> str: 

        if frequencia_hz < 120.0:
            classificacao = "PROPULSAO_PESADA"
        elif 120.0 <= frequencia_hz <= 800.0: # elif (frequencia_hz >= 120.0) and (frequencia_hz <= 800.0):
            classificacao = "SUBMARINO_DIESEL"
        else: # frequencia_hz > 800.0
            classificacao = "BIOLOGICO"

        contado_id = contato_id.strip().upper()
        self.assinaturas_frequencia[contato_id] = classificacao 
        return classificacao

        

    # Sobrescreva registrar_contato(self, contato_id: str, tipo: str, distancia_km: float, 
    # frequencia_hz: Optional[float] = None) -> bool:
    # Chama a versão da superclasse via super().registrar_contato(...).
    # Se o registro for bem-sucedido e frequencia_hz is not None, invoca self.analisar_frequencia(contato_id, frequencia_hz) 
    # e adiciona a chave "assinatura" no dicionário do contato dentro de self.contatos_detectados.
    # Retorna o booleano de sucesso.
    def registrar_contato(self , 
                          contato_id    : str, 
                          tipo          : str, 
                          distancia_km  : float, 
                          frequencia_hz : float = None
                         ) -> bool:

        ok = SensorAcustico.registrar_contato(self, contato_id, tipo, distancia_km )  # True or False
        print(ok)
        print(self.contatos_detectados)
        if ok and frequencia_hz !=None:
            # tentativa de classificacao
            assinatura = self.analisar_frequencia(contato_id, frequencia_hz) 
            self.contatos_detectados[contato_id]["assinatura"] = assinatura
            print(self.contatos_detectados)
            return True

        return False

# Teste do Exercício 5:
# sonar = SonarPassivoAvancado(nome_sensor="Sonar-CIAA-Pro", alcance_km=50.0, sensibilidade_db=95.5)
# sonar.registrar_contato("ct-01", "Desconhecido", distancia_km=32.0, frequencia_hz=85.0)
# sonar.registrar_contato("ct-02", "Baleia", distancia_km=12.0, frequencia_hz=1200.0)
# sonar.registrar_contato("ct-03", "Navio", distancia_km=60.0) # Fora de alcance
# print("Contatos:", sonar.contatos_detectados)
# print("Assinaturas:", sonar.assinaturas_frequencia)
```

### Exercício 6: Computador de Navegação e Auditoria de Rotas Táticas (Herança + Listas de Dicionários + Validações)

**Cenário**: O submarino planeja rotas de navegação submarina através de uma lista de pontos geográficos (*waypoints*). Para missões sigilosas, o computador deve auditar automaticamente se algum trecho trafega em cotas de profundidade perigosamente rasas.

**Instruções**:
1. **Superclasse `ComputadorNavegacao`**:
   * `__init__(self, codigo_submarino: str) -> None`: Armazena `self.codigo_submarino: str` e `self.waypoints: List[Dict[str, Union[str, float]]] = []`.
   * `adicionar_waypoint(self, nome_ponto: str, profundidade_metros: float, distancia_nm: float) -> bool`:
     1. Valida se `profundidade_metros >= 0.0` e `distancia_nm > 0.0`. Se inválido, imprime aviso e retorna `False`.
     2. Formata o nome do ponto com `nome_ponto.strip().title()`.
     3. Appenda o dicionário `{"ponto": nome_formatado, "profundidade": profundidade_metros, "distancia_nm": distancia_nm}` em `self.waypoints` e retorna `True`.
   * `calcular_distancia_total(self) -> float`: Retorna a soma de todas as distâncias náuticas (`distancia_nm`) dos waypoints cadastrados.

2. **Subclasse `ComputadorTaticoSigiloso(ComputadorNavegacao)`**:
   * Herda de `ComputadorNavegacao`.
   * `__init__(self, codigo_submarino: str, profundidade_segura_minima: float) -> None`: Inicializa com `super()` e adiciona `self.profundidade_segura_minima: float = profundidade_segura_minima` e `self.alertas_seguranca: List[str] = []`.
   * Sobrescreva o método `adicionar_waypoint(self, nome_ponto: str, profundidade_metros: float, distancia_nm: float) -> bool`:
     * Chama `super().adicionar_waypoint(...)`.
     * Se inserido com sucesso e `profundidade_metros < self.profundidade_segura_minima`, imprime:
       `f"⚠️ ALERTA TÁTICO: Waypoint '{nome_ponto.strip().title()}' em cota rasa ({profundidade_metros:.1f}m < {self.profundidade_segura_minima:.1f}m)!"`.
     * Retorna o booleano de sucesso.
   * Método `auditar_rota_completa(self) -> Dict[str, Union[int, float, List[str]]]`:
     * Limpa e reconstrói `self.alertas_seguranca`.
     * Para cada waypoint com `profundidade < self.profundidade_segura_minima`, adiciona uma mensagem explicativa em `self.alertas_seguranca`.

       * `"vulnerabilidades_detectadas"`: quantidade de alertas (int)
       * `"alertas"`: lista de strings com os alertas (List[str])

* **Exigência**: Utilize anotações de tipo (**Type Hints**) completas (`Dict`, `List`, `Union`, `bool`, `float`, `str`, `None`).

```python
# IMPLEMENTAÇÃO DO EXERCÍCIO 6
from typing import List, Dict, Union

# 1. Superclasse ComputadorNavegacao
class ComputadorNavegacao:
    
    #self, codigo_submarino: str) -> None: Armazena self.codigo_submarino: str e 
    #self.waypoints: List[Dict[str, Union[str, float]]] = [].
    def __init__(self, codigo_submarino : str ) -> None:

        self.codigo_submarino= codigo_submarino 
        self.waypoints = [] #self.waypoints: List[Dict[str, Union[str, float]]] = [].


    # adicionar_waypoint(self, nome_ponto: str, profundidade_metros: float, distancia_nm: float) -> bool:
    # Valida se profundidade_metros >= 0.0 e distancia_nm > 0.0. Se inválido, imprime aviso e retorna False.
    # Formata o nome do ponto com nome_ponto.strip().title().
    # Appenda o dicionário {"ponto": nome_formatado, "profundidade": profundidade_metros, "distancia_nm": distancia_nm} 
    # em self.waypoints e retorna True.
    def adicionar_waypoint(self, 
                           nome_ponto: str, 
                           profundidade_metros: float, 
                           distancia_nm: float
                          ) -> bool:

        if profundidade_metros < 0.0:
            print("profundidade deve ser maior ou igual a zero")
            return False

        if distancia_nm < 0.0:
            print("distancia_nm deve ser maior ou igual a zero")
            return False

        nome_ponto = nome_ponto.strip().title() # joao -> Joao , JOAO -> Joao?

        ponto = {"ponto": nome_ponto, "profundidade" : profundidade_metros, "distancia_nm": distancia_nm}
        self.waypoints.append(ponto)
        return True
        
    def calcular_distancia_total(self) -> float: 
        distancia_total = 0
        for ponto in self.waypoints: # [ ponto1, ponto2, ponto3] onde ponto1
            distancia_total += ponto["distancia_nm"]
        return distancia_total

        

        

# 2. Subclasse ComputadorTaticoSigiloso
class ComputadorTaticoSigiloso(ComputadorNavegacao):

    def __init__(self, codigo_submarino: str, profundidade_segura_minima: float) -> None:

        ComputadorNavegacao.__init__(self, codigo_submarino )
        self.profundidade_segura_minima: float = profundidade_segura_minima 
        self.alertas_seguranca: List[str] = []

        
    #Sobrescreva o método adicionar_waypoint(self, nome_ponto: str, profundidade_metros: float, distancia_nm: float) -> bool:
    #Chama super().adicionar_waypoint(...).
    #Se inserido com sucesso e profundidade_metros < self.profundidade_segura_minima, 
    # imprime: f"⚠️ ALERTA TÁTICO: Waypoint '{nome_ponto.strip().title()}' em cota 
    # rasa ({profundidade_metros:.1f}m < {self.profundidade_segura_minima:.1f}m)!".
    #Retorna o booleano de sucesso.

    def adicionar_waypoint(self, nome_ponto: str, profundidade_metros: float, distancia_nm: float) -> bool:

        ok = ComputadorNavegacao.adicionar_waypoint( self,  nome_ponto, profundidade_metros, distancia_nm)

        if ok and (profundidade_metros < self.profundidade_segura_minima):
            prof = round(profundidade_metros, 1)
            print(f"⚠️ ALERTA TÁTICO: Waypoint '{nome_ponto.strip().title()}' em cota rasa ({prof}m < {self.profundidade_segura_minima}m)!")
            return True

        return False


    
    # Método auditar_rota_completa(self) -> Dict[str, Union[int, float, List[str]]]:
    # Limpa e reconstrói self.alertas_seguranca.
    # Para cada waypoint com profundidade < self.profundidade_segura_minima, adiciona uma mensagem explicativa 
    # em self.alertas_seguranca.
    # Retorna um dicionário resumo:
    #"total_pontos": len(self.waypoints) (int)
    # "distancia_total_nm": self.calcular_distancia_total() (float)
    # "vulnerabilidades_detectadas": quantidade de alertas (int)
    # "alertas": lista de strings com os alertas (List[str])

    def auditar_rota_completa(self) -> Dict[str, Any]:

     

        for ponto in self.waypoints:

            profundidade = ponto["profundidade"]
            nome_ponto = ponto["ponto"]
            if profundidade < self.profundidade_segura_minima:
                self.alertas_seguranca.append( f"ALERTA {nome_ponto}")

        sumario = {
            "total_pontos": len(self.waypoints),
            "distancia_total_nm": self.calcular_distancia_total(),
            "vulnerabilidades_detectadas": len(self.alertas_seguranca),
            "alertas" : self.alertas_seguranca,
        }

        return sumario


# Teste do Exercício 6:
nav = ComputadorTaticoSigiloso(codigo_submarino="CIAA-LPS-Alpha", profundidade_segura_minima=100.0)
nav.adicionar_waypoint("fossa das marianas - setor a", profundidade_metros=350.0, distancia_nm=45.0)
nav.adicionar_waypoint("canal de recifes costeiros", profundidade_metros=40.0, distancia_nm=22.5) # Alerta cota rasa
nav.adicionar_waypoint("plataforma continental", profundidade_metros=80.0, distancia_nm=30.0)     # Alerta cota rasa
```
