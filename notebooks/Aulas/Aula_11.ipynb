### 🛒 Exercício Prático 1.1: Cálculo de Consumo e Custo de Viagem de Carro

**Contexto e Regras de Negócio:**  
Você está planejando uma viagem rodoviária de férias (ou calculando os custos operacionais de uma corrida por aplicativo). O sistema precisa de uma função que receba a distância total a ser percorrida, o rendimento médio do veículo e o preço do combustível na bomba, calculando o volume de combustível exigido e o gasto financeiro total da viagem.

**Assinatura da Função:**
```python
def calcular_custo_viagem(distancia_km: float, consumo_km_l: float, preco_combustivel: float) -> tuple[float, float]:
```

**Parâmetros de Entrada:**
* `distancia_km` (`float`): Distância total do trajeto em quilômetros (ex: `420.0`).
* `consumo_km_l` (`float`): Autonomia média do veículo em km por litro (ex: `12.5`).
* `preco_combustivel` (`float`): Preço unitário do litro de combustível em R$ (ex: `5.89`).

**Processamento:**
1. $	ext{Litros Necessários} = rac{	ext{distancia\_km}}{	ext{consumo\_km\_l}}$
2. $	ext{Custo Total (R\$)} = 	ext{Litros Necessários} 	imes 	ext{preco\_combustivel}$

**Formato da Saída e Retorno Esperado:**
* Retornar uma **tupla** de dois floats: `(litros_necessarios, custo_total)` com ambos os valores arredondados para 2 casas decimais (`round(valor, 2)`).
* *Exemplo com `(420.0, 12.5, 5.89)`*: Retorna `(33.6, 197.9)`.


````python
# ==============================================================================
# EXERCÍCIO 1.1 - ESPAÇO PARA IMPLEMENTAÇÃO
# ==============================================================================

def calcular_custo_viagem(
    distancia_km: float, 
    consumo_km_l: float, 
    preco_combustivel: float
) -> tuple[float, float]:
    """
    Calcula os litros de combustível necessários e o custo total de uma viagem.
    """

    # quantidade de combustivel
    litros = distancia_km / consumo_km_l

    # custo 
    custo_total = preco_combustivel * litros

    #return (litros, custto_total) # tupla
    return litros, custo_total # tupla por default
    


# ------------------------------------------------------------------------------
# Teste o seu código abaixo:
# ------------------------------------------------------------------------------
dist = 420.0       # 420 km
consumo = 12.5     # 12.5 km/l
preco_litro = 5.89 # R$ 5.89 por litro

litros, custo = calcular_custo_viagem(dist, consumo, preco_litro)
print(f"Combustível necessário: {litros:.2f} litros | Custo: R$ {custo:.2f}")

# Saída Esperada:
# Combustível necessário: 33.60 litros | Custo: R$ 197.90
````

### 🏠 Exercício Prático 2.2: Validação de Análise de Crédito para Financiamento

**Contexto e Regras de Negócio:**  
Um aplicativo bancário de empréstimos e financiamentos imobiliários/veiculares avalia automaticamente as propostas submetidas pelos clientes. Para que um financiamento seja aprovado, o cliente deve cumprir simultaneamente todos os seguintes critérios:
1. **Faixa Etária Permitida**: A idade do solicitante deve estar entre **18 e 75 anos** (inclusive).
2. **Histórico Cadastral**: O solicitante **não pode possuir restrições no CPF** (`possui_restricao == False`).
3. **Capacidade de Pagamento**: O valor da parcela mensal solicitada **não pode ultrapassar 30% da renda líquida mensal** do cliente ($	ext{valor\_parcela} \le 0.30 	imes 	ext{renda\_mensal}$).

**Assinatura da Função:**
```python
def analisar_financiamento(idade: int, renda_mensal: float, valor_parcela: float, possui_restricao: bool) -> tuple[bool, str]:
```

**Parâmetros de Entrada:**
* `idade` (`int`): Idade do cliente em anos (ex: `28`).
* `renda_mensal` (`float`): Renda líquida comprovada em R$ (ex: `6000.0`).
* `valor_parcela` (`float`): Parcela mensal desejada em R$ (ex: `1500.0`).
* `possui_restricao` (`bool`): Indicador de negativação no SPC/Serasa (`True` ou `False`).

**Formato da Saída e Retorno Esperado:**
* Retornar uma **tupla** de dois elementos: `(aprovado: bool, justificativa: str)`.
  * Se aprovado: `(True, "Aprovado: Proposta de crédito atende a todos os critérios operacionais!")`.
  * Se reprovado por idade: `(False, "Reprovado: Idade (X anos) fora da faixa permitida (18 a 75 anos).")`.
  * Se reprovado por restrição: `(False, "Reprovado: CPF com restrições cadastrais nos órgãos de proteção ao crédito.")`.
  * Se reprovado por parcela alta: `(False, "Reprovado: Parcela de R$ X excede o limite de 30% da renda (máx: R$ Y).")`.
* *Exemplo com `(28, 6000.0, 1500.0, False)`*: Retorna `(True, "Aprovado: Proposta de crédito atende a todos os critérios operacionais!")`.


````python
# ==============================================================================
# EXERCÍCIO 2.2 - ESPAÇO PARA IMPLEMENTAÇÃO
# ==============================================================================

def analisar_financiamento(
    idade: int, 
    renda_mensal: float, 
    valor_parcela: float, 
    possui_restricao: bool
) -> tuple[bool, str]:
    """
    Valida critérios de aprovação de crédito bancário.
    """

    renda_liquida_mensal = renda_mensal
    
    aceita = True
    status_code = "APROVADO"
    
    if idade < 18 or idade > 75:
        aceita = False
        status_code = "REPROVADO_IDADE"
        return aceita, status_code

    
    if valor_parcela > (0.3*renda_liquida_mensal):
        aceita = False
        status_code = "REPROVADO_PARCELA_SUPERIOR_30"
        return aceita, status_code

    if possui_restricao: # possui_restricao -> True
        aceita = False
        status_code = "REPROVADO_SERASA"

    if valor_parcela < 0:
        aceita = False
        status_code = "REPROVADO_PARCELA_MENOR_ZERO"
        return aceita, status_code

    return aceita, status_code
    



# ------------------------------------------------------------------------------
# Teste o seu código abaixo:
# ------------------------------------------------------------------------------
aprovado, msg = analisar_financiamento(8, 6000.0, 1000.0, True)
print(f"Aprovado: {aprovado} | Motivo: {msg}")

# Saída Esperada:
# Aprovado: True | Motivo: Aprovado: Proposta de crédito atende a todos os critérios operacionais!
````

### 💰 Exercício Prático 3.1: Meta Financeira de Poupança com Juros Mensais (`while`)

**Contexto e Regras de Negócio:**  
Você deseja comprar um bem ou fazer uma viagem que custa um determinado valor alvo (ex: **R$ 10.000,00**). Você abre uma aplicação com um saldo inicial (ex: R$ 1.000,00) e todo mês deposita uma quantia fixa (ex: R$ 600,00). O dinheiro na conta rende juros mensais compostos (ex: **0.8% ao mês**) sobre o saldo acumulado.

**Dinâmica de cada mês (no laço `while`):**
1. O saldo acumulado anterior rende a taxa de juros do mês: $	ext{saldo} = 	ext{saldo} + (	ext{saldo} 	imes 	ext{taxa\_juros\_mensal})$.
2. Em seguida, o aporte fixo mensal é adicionado: $	ext{saldo} = 	ext{saldo} + 	ext{deposito\_mensal}$.
3. O contador de meses é incrementado em $+1$.
4. O laço se repete até que $	ext{saldo} \ge 	ext{meta\_final}$.

**Assinatura da Função:**
```python
def calcular_tempo_meta_poupanca(saldo_inicial: float, deposito_mensal: float, meta_final: float, taxa_juros_mensal: float = 0.008) -> tuple[int, float]:
```

**Parâmetros de Entrada:**
* `saldo_inicial` (`float`): Valor inicial depositado em R$ (ex: `1000.0`).
* `deposito_mensal` (`float`): Aporte fixo realizado a cada mês em R$ (ex: `600.0`).
* `meta_final` (`float`): Montante financeiro pretendido em R$ (ex: `10000.0`).
* `taxa_juros_mensal` (`float`, padrão `0.008` = 0.8% a.m.).

**Formato da Saída e Retorno Esperado:**
* Retornar uma **tupla**: `(meses_necessarios: int, saldo_final_acumulado: float)` com o saldo arredondado para 2 casas decimais.
* *Exemplo com `(1000.0, 600.0, 10000.0, 0.008)`*: Retorna `(15, 10648.86)`.


````python
# ==============================================================================
# EXERCÍCIO 3.1 - ESPAÇO PARA IMPLEMENTAÇÃO
# ==============================================================================

def calcular_tempo_meta_poupanca(
    saldo_inicial: float, 
    deposito_mensal: float, 
    meta_final: float, 
    taxa_juros_mensal: float = 0.008
) -> tuple[int, float]:
    """
    Calcula quantos meses são necessários para atingir uma meta financeira com rendimentos via while.
    """

    saldo = saldo_inicial # v0
    meses = 0

    while saldo < meta_final:

        meses += 1
        saldo = saldo + taxa_juros_mensal*saldo
        saldo = saldo + deposito_mensal
    
    return meses, saldo


# ------------------------------------------------------------------------------
# Teste o seu código abaixo:
# ------------------------------------------------------------------------------
meses, saldo_final = calcular_tempo_meta_poupanca(1000.0, 600.0, 10000.0, 0.008)
print(f"Meses necessários: {meses} | Saldo final acumulado: R$ {saldo_final:.2f}")

# Saída Esperada:
# Meses necessários: 15 | Saldo final acumulado: R$ 10648.86
````

````python
# ==============================================================================
# EXERCÍCIO 4.1 - ESPAÇO PARA IMPLEMENTAÇÃO
# ==============================================================================

def formatar_dados_cliente(nome_completo: str, telefone_bruto: str) -> dict:
    """
    Sanitiza nome e telefone de formulário web retornando dados estruturados.
    """

    nome_completo = nome_completo.strip()
    partes_do_nome = nome_completo.split()
    nome_padronizado = " ".join(partes_do_nome).title()

    primeiro_nome = partes_do_nome[0].title()

    ultimo_nome = partes_do_nome[-1].title()
    telefone_limpo = ""
    
    for c in telefone_bruto:
        if c.isdigit():
            telefone_limpo += c

    if len(telefone_limpo) == 11:
        print("aqui joao")
        ddd = telefone_limpo[0:2] # 0,1
        parte_1 = telefone_limpo[2:7] # 2 até 6
        parte_2 = telefone_limpo[7::] # 8 até o final, 8:11
        telefone_formatado = f"({ddd}) {parte_1}-{parte_2}"
    else:
        telefone_formatado = telefone_limpo

    return {
        "nome_padronicado" : nome_padronizado,
        "primeiro_nome" : primeiro_nome,
        "ultimo_nome" : ultimo_nome,
        "telefone" : telefone_formatado,
    }


# ------------------------------------------------------------------------------
# Teste o seu código abaixo:
# ------------------------------------------------------------------------------
res = formatar_dados_cliente("   joao victor  fonseca pinto  ", "(21) 98765-4321")
print(res)

# Saída Esperada:
# {'nome_padronizado': 'Joao Victor Fonseca Pinto', 'primeiro_nome': 'Joao', 'ultimo_sobrenome': 'Pinto', 'telefone': '(21) 98765-4321'}
````

### 🏃 Exercício Prático 4.2: Distância Total de Caminhada no Parque com Tuplas GPS

**Contexto e Regras de Negócio:**  
Um aplicativo de corrida e condicionamento físico (como Strava ou Nike Run) rastreia a atividade do atleta gravando uma sequência de pontos de posicionamento no espaço 2D. Cada ponto é modelado como uma **tupla imutável `(x, y)`** com as coordenadas cartesianas em **metros**.

**Cálculo Geométrico da Rota:**
* Para cada segmento consecutivo entre o ponto $P_i(x_i, y_i)$ e o ponto $P_{i+1}(x_{i+1}, y_{i+1})$, a distância euclidiana em metros é dada por:
  $$d_{	ext{segmento}} = \sqrt{(x_{i+1} - x_i)^2 + (y_{i+1} - y_i)^2}$$
* A distância total do treino é a soma de todos os segmentos convertida de **metros para quilômetros** ($	ext{distancia\_km} = rac{\sum d_{	ext{segmento}}}{1000.0}$).

**Assinatura da Função:**
```python
def calcular_distancia_treino(trajeto: list[tuple[float, float]]) -> float:
```

**Parâmetros de Entrada:**
* `trajeto` (`list[tuple[float, float]]`): Lista de tuplas com as coordenadas `(x, y)` em metros (ex: `[(0.0, 0.0), (500.0, 0.0), (500.0, 800.0), (1200.0, 800.0), (1200.0, 0.0), (0.0, 0.0)]`).

**Formato da Saída e Retorno Esperado:**
* Retornar um `float` representando a distância total percorrida em **quilômetros**, arredondada para 3 casas decimais (`round(dist_km, 3)`). Se a lista possuir menos de 2 pontos, retornar `0.0`.
* *Exemplo com a rota acima*: Retorna `4.0` (ou `4.000` km).


````python
# ==============================================================================
# EXERCÍCIO 4.2 - ESPAÇO PARA IMPLEMENTAÇÃO
# ==============================================================================


def calcular_distancia_treino(trajeto: list[tuple[float, float]]) -> float:
    """
    Calcula a distância total percorrida em um treino em quilômetros a partir de tuplas (x, y).
    """
    dist_total_metros = 0
    # P1 -> P2 = D12
    # P2 -> P3 = D23
    # P_n-1 -> P_n
    for trajeto_i in range(len(trajeto) - 1):
        x1, y1 = trajeto[trajeto_i] # n
        x2, y2 = trajeto[trajeto_i + 1] # n + 1
        dist_segmento = ( (x2-x1)**2 + (y2-y1)**2 )**(0.5) # math.sqrt( ((x2-x1)**2 + (y2-y1)**2) )
        dist_total_metros += dist_segmento

    dist_total_km = dist_total_metros / 1000
    return dist_total_km

    


# ------------------------------------------------------------------------------
# Teste o seu código abaixo:
# ------------------------------------------------------------------------------
pontos = [
     (0.0, 0.0), (500.0, 0.0), (500.0, 800.0), 
     (1200.0, 800.0), (1200.0, 0.0), (0.0, 0.0)
]
km = calcular_distancia_treino(pontos)
print(f"Distância total: {km:.3f} km")

# Saída Esperada:
# Distância total: 4.000 km
````

### 🍽️ Exercício Prático 5.1: Média Aparada de Avaliações de Restaurante (Remoção de Notas Extremas)

**Contexto e Regras de Negócio:**  
Em plataformas gastronômicas e de delivery (como iFood ou TripAdvisor), a reputação de um estabelecimento pode ser injustamente distorcida por avaliações maliciosas (nota 1.0) ou por notas artificiais de bots (nota 5.0). Para obter uma métrica confiável e justa, calcula-se a **Média Aparada** (*Trimmed Mean*), que descarta uma porcentagem de corte dos menores e dos maiores valores da amostra antes de calcular a média aritmética.

**Requisitos de Implementação:**
1. **Proteção de Dados**: A lista original de notas do restaurante **não pode ser alterada** pela função. Trabalhe obrigatoriamente sobre uma cópia ordenada independente (`sorted(notas)`).
2. Determinar a quantidade de itens a serem descartados em cada extremidade: $	ext{qtd\_corte} = 	ext{int}(N 	imes 	ext{pct\_corte})$.
3. Fatiar a lista eliminando os `qtd_corte` primeiros e os `qtd_corte` últimos elementos (se $2 	imes 	ext{qtd\_corte} < N$).
4. Calcular a média dos valores centrais remanescentes.

**Assinatura da Função:**
```python
def calcular_avaliacao_justa(notas: list[float], pct_corte: float = 0.1) -> float:
```

**Parâmetros de Entrada:**
* `notas` (`list[float]`): Lista com as notas de 0 a 5 dos clientes (ex: `[1.0, 4.5, 4.8, 4.6, 4.7, 4.9, 4.5, 4.8, 5.0, 5.0]`).
* `pct_corte` (`float`, padrão `0.1` = 10% de corte em cada ponta).

**Formato da Saída e Retorno Esperado:**
* Retornar um `float` com a média aparada calculada, arredondada para 2 casas decimais (`round(media, 2)`). Caso a lista de notas seja vazia, retornar `0.0`.
* *Exemplo com as notas acima e corte de 10%*: Retorna `4.73`.


````python
# ==============================================================================
# EXERCÍCIO 5.1 - ESPAÇO PARA IMPLEMENTAÇÃO
# ==============================================================================

def calcular_avaliacao_justa(notas: list[float], pct_corte: float = 0.1) -> float:
    """
    Calcula a média de notas de um restaurante descartando os extremos (outliers).
    """
    notas_ordenadas = sorted(notas)
    n = len(notas_ordenadas)
    qtd_corte = int(n*pct_corte)

    if qtd_corte > 0 and (qtd_corte*2 < n):
        # [0, 1, 2 , n_cut, ....... n - n_cut, n - n_cut -1 , ... n]
        notas_filtradas = notas_ordenadas[  qtd_corte : n - qtd_corte  ]
    else:
        notas_filtradas = notas_ordenadas


    media_aparada = sum(notas_filtradas) / len(notas_filtradas)

    return media_aparada


# ------------------------------------------------------------------------------
# Teste o seu código abaixo:
# ------------------------------------------------------------------------------
avaliacoes = [1.0, 4.5, 4.8, 4.6, 4.7, 4.9, 4.5, 4.8, 5.0, 5.0]
media_aparada = calcular_avaliacao_justa(avaliacoes, pct_corte=0.1)
print(f"Média Aparada Justa: {media_aparada:.2f}")

# Saída Esperada:
# Média Aparada Justa: 4.73
````

### 🍕 Exercício Prático 6.1: Sistema de Comanda e Conta de Restaurante/Pizzaria (Dicionários)

**Contexto e Regras de Negócio:**  
Em um restaurante ou pizzaria, o sistema gerencia o cardápio oficial através de um dicionário de preços `{nome_item: preco_unitario}` e recebe a comanda de consumo de uma mesa com as quantidades pedidas `{nome_item: quantidade_consumida}`.

**Processamento Requerido:**
1. Para cada item consumido, buscar seu preço unitário no cardápio via `.get(item, 0.0)` e calcular o valor do item: $	ext{total\_item} = 	ext{preco\_unitario} 	imes 	ext{quantidade}$.
2. Somar o **subtotal** da mesa.
3. Calcular a **taxa de serviço** (padrão de 10%): $	ext{taxa\_servico} = 	ext{subtotal} 	imes \left(rac{	ext{taxa\_servico\_pct}}{100}ight)$.
4. Calcular o **total a pagar**: $	ext{total\_geral} = 	ext{subtotal} + 	ext{taxa\_servico}$.

**Assinatura da Função:**
```python
def fechar_conta_mesa(cardapio: dict[str, float], consumo: dict[str, int], taxa_servico_pct: float = 10.0) -> dict:
```

**Parâmetros de Entrada:**
* `cardapio` (`dict[str, float]`): Tabela de preços `{produto: preco}`.
* `consumo` (`dict[str, int]`): Pedidos da mesa `{produto: quantidade}`.
* `taxa_servico_pct` (`float`, padrão `10.0` = 10%).

**Formato da Saída e Retorno Esperado:**
* Retornar um **dicionário (`dict`)** com:
  * `"itens"` (`list[dict]`): Lista com o extrato detalhado de cada produto: `{"item": str, "quantidade": int, "preco_unitario": float, "total_item": float}`.
  * `"subtotal"` (`float`): Subtotal somado dos produtos consumidos em R$.
  * `"taxa_servico"` (`float`): Valor financeiro da taxa de serviço calculada em R$.
  * `"total_pagar"` (`float`): Valor total da conta a ser cobrado do cliente em R$.
* Todos os floats devem ser arredondados para 2 casas decimais.
* *Exemplo com consumo de R$ 183.00*: Retorna subtotal R$ 183.00, taxa R$ 18.30 e total R$ 201.30.


````python
# ==============================================================================
# EXERCÍCIO 6.1 - ESPAÇO PARA IMPLEMENTAÇÃO
# ==============================================================================

def fechar_conta_mesa(
    cardapio: dict[str, float], 
    consumo: dict[str, int], 
    taxa_servico_pct: float = 10.0
) -> dict:
    """
    Calcula o extrato detalhado e o valor final da comanda de uma mesa.
    """

    total_conta = 0
    detalhes_do_pedido = []

    
    for item_consumido, qtd in consumo.items():

        preco_unit = cardapio.get( item_consumido, 0 )
        total_item = preco_unit * qtd

        total_conta += total_item

        detalhe_do_item = {
            "item" : item_consumido , # pizza ou refrigente, etc
            "quantidade" : qtd,
            "preco_unitario" : preco_unit,
            "total_item" : total_item
        }

        detalhes_do_pedido.append( detalhe_do_item)


    valor_taxa = total_conta * taxa_servico_pct/100
    total = total_conta + valor_taxa

    return {
        "detalhes_do_pedido" : detalhes_do_pedido,
        "subtotal" :  total_conta,
        "taxa_servico" : valor_taxa,
        "total_pagar" : total,
    }
    


# ------------------------------------------------------------------------------
# Teste o seu código abaixo:
# ------------------------------------------------------------------------------
cardapio = {"Pizza Margherita": 55.00, "Refrigerante 2L": 12.00}
consumo = {"Pizza Margherita": 1, "Refrigerante 2L": 2}
conta = fechar_conta_mesa(cardapio, consumo, taxa_servico_pct=10.0)

print(conta)
# print(f"Subtotal: R$ {conta['subtotal']:.2f} | Total: R$ {conta['total_pagar']:.2f}")

# Saída Esperada:
# Subtotal: R$ 79.00 | Total: R$ 86.90
````

````python
# ==============================================================================
# EXERCÍCIO 6.2 - SOLUÇÃO COMENTADA
# ==============================================================================

def analisar_compatibilidade_musical(
    artistas_usuario_a: list[str], 
    artistas_usuario_b: list[str]
) -> dict:
    set_a = set(artistas_usuario_a)
    set_b = set(artistas_usuario_b)
    
    # 1. Interseção (&): Artistas ouvidos por ambos
    em_comum = set_a & set_b
    
    # 2. Diferença (-): Artistas de B que A não ouve
    sugestoes_para_a = set_b - set_a
    
    # 3. Diferença (-): Artistas de A que B não ouve
    sugestoes_para_b = set_a - set_b
    
    # 4. União (|): Todos os artistas do catálogo combinado
    catalogo_total = set_a | set_b
    
    afinidade_pct = (len(em_comum) / len(catalogo_total) * 100.0) if catalogo_total else 0.0
    
    return {
        "artistas_em_comum": sorted(list(em_comum)),
        "recomendacoes_para_usuario_a": sorted(list(sugestoes_para_a)),
        "recomendacoes_para_usuario_b": sorted(list(sugestoes_para_b)),
        "afinidade_percentual": round(afinidade_pct, 1)
    }

# Execução e Validação:
ana_artistas = ["Coldplay", "Queen", "Imagine Dragons", "Dua Lipa", "The Beatles", "Queen"]
carlos_artistas = ["Queen", "The Beatles", "Linkin Park", "Foo Fighters", "Coldplay"]
resultado_afinidade = analisar_compatibilidade_musical(ana_artistas, carlos_artistas)
print(f"🎧 Afinidade: {resultado_afinidade['afinidade_percentual']}%")
print(f"   🤝 Em Comum: {resultado_afinidade['artistas_em_comum']}")
print(f"   💡 Sugestões para Ana: {resultado_afinidade['recomendacoes_para_usuario_a']}")
print(f"   💡 Sugestões para Carlos: {resultado_afinidade['recomendacoes_para_usuario_b']}")
````

   * **Método `alugar(self) -> bool`**:

   * **Método `devolver(self, dias: int) -> float`**:

   * **Método Especial `__repr__(self) -> str`**:

2. **Classe Derivada `CarroEletrico(Veiculo)`** (Herda de `Veiculo`):
   * **Construtor `__init__(self, modelo: str, ano: int, valor_diaria: float, capacidade_bateria_kwh: float)`**:
     * Invoca o construtor pai via `super().__init__(modelo, ano, valor_diaria)`.

   * **Método `recarregar(self)`**:

   * **Método `dirigir(self, km: float) -> bool`**:

````python
# ==============================================================================
# EXERCÍCIO 7.1 - SOLUÇÃO COMENTADA
# ==============================================================================

class Veiculo:
    def __init__(self, modelo: str, ano: int, valor_diaria: float):
        self.modelo = modelo
        self.ano = ano
        self.valor_diaria = valor_diaria
        self.alugado = False
        
    def alugar(self) -> bool:
        if self.alugado:
            print(f"⚠️ [{self.modelo}] Já está alugado.")
            return False
        self.alugado = True
        print(f"🔑 [{self.modelo}] Aluguel iniciado! Diária: R$ {self.valor_diaria:.2f}")
        return True
        
    def devolver(self, dias: int) -> float:
        if not self.alugado:
            print(f"⚠️ [{self.modelo}] Não estava alugado.")
            return 0.0
        self.alugado = False
        total = dias * self.valor_diaria
        print(f"🏁 [{self.modelo}] Devolvido após {dias} dias. Total: R$ {total:.2f}")
        return total
        
    def __repr__(self) -> str:
        status = "Alugado" if self.alugado else "Disponível"
        return f"<Veiculo: {self.modelo} ({self.ano}) | R$ {self.valor_diaria:.2f}/dia | {status}>"


class CarroEletrico(Veiculo):
    def __init__(self, modelo: str, ano: int, valor_diaria: float, capacidade_bateria_kwh: float):
        # Chama o construtor da classe pai com super()
        super().__init__(modelo, ano, valor_diaria)
        self.capacidade_bateria_kwh = capacidade_bateria_kwh
        self.nivel_bateria_pct = 100.0
        
    def recarregar(self):
        self.nivel_bateria_pct = 100.0
        print(f"⚡ [{self.modelo}] Bateria 100% recarregada!")
        
    def dirigir(self, km: float) -> bool:
        consumo_pct = km * 0.25
        if consumo_pct > self.nivel_bateria_pct:
            print(f"🚨 [{self.modelo}] Carga insuficiente para {km} km.")
            return False
        self.nivel_bateria_pct -= consumo_pct
        print(f"🛣️ [{self.modelo}] Viagem de {km} km concluída. Bateria: {self.nivel_bateria_pct:.1f}%")
        return True
        
    def __repr__(self) -> str:
        status = "Alugado" if self.alugado else "Disponível"
        return f"<Carro Elétrico: {self.modelo} ({self.ano}) | R$ {self.valor_diaria:.2f}/dia | Bateria: {self.nivel_bateria_pct:.1f}% | {status}>"

# Execução e Validação:
carro_byd = CarroEletrico("BYD Dolphin", 2024, 220.0, 44.9)
carro_byd.alugar()
carro_byd.dirigir(100.0)
total_pago = carro_byd.devolver(dias=2)
print(carro_byd, f"| Total a pagar: R$ {total_pago:.2f}")
````

````python
# ==============================================================================
# EXERCÍCIO 7.2 - SOLUÇÃO COMENTADA
# ==============================================================================

import os

def salvar_lancamentos_financeiros(caminho_arquivo: str, lancamentos: list[dict]) -> None:
    # 1. Utiliza o context manager 'with open' para fechamento automático
    with open(caminho_arquivo, "w", encoding="utf-8") as f:
        # Grava o cabeçalho CSV
        f.write("data,descricao,categoria,tipo,valor\n")
        for l in lancamentos:
            linha = f"{l['data']},{l['descricao']},{l['categoria']},{l['tipo']},{l['valor']:.2f}\n"
            f.write(linha)
    print(f"💾 {len(lancamentos)} lançamentos salvos com sucesso em '{caminho_arquivo}'.")


def carregar_e_calcular_saldo(caminho_arquivo: str) -> dict:
    if not os.path.exists(caminho_arquivo):
        raise FileNotFoundError(f"Arquivo '{caminho_arquivo}' não encontrado.")
        
    total_receitas = 0.0
    total_despesas = 0.0
    lista_itens = []
    
    # 2. Abre o arquivo em modo de leitura
    with open(caminho_arquivo, "r", encoding="utf-8") as f:
        linhas = f.readlines()
        
    # Pula a linha 0 (cabeçalho)
    for linha in linhas[1:]:
        linha_limpa = linha.strip()
        if not linha_limpa:
            continue
        partes = linha_limpa.split(",")
        if len(partes) == 5:
            dt, desc, cat, tipo, val_str = partes
            val = float(val_str)
            if tipo == "RECEITA":
                total_receitas += val
            elif tipo == "DESPESA":
                total_despesas += val
                
            lista_itens.append({
                "data": dt,
                "descricao": desc,
                "categoria": cat,
                "tipo": tipo,
                "valor": val
            })
            
    saldo_final = total_receitas - total_despesas
    
    return {
        "total_receitas": round(total_receitas, 2),
        "total_despesas": round(total_despesas, 2),
        "saldo_liquido": round(saldo_final, 2),
        "quantidade_lancamentos": len(lista_itens)
    }

# Execução e Validação:
dados_teste = [
    {"data": "01/10/2026", "descricao": "Salário", "categoria": "Trabalho", "tipo": "RECEITA", "valor": 5000.00},
    {"data": "05/10/2026", "descricao": "Aluguel", "categoria": "Moradia", "tipo": "DESPESA", "valor": 1500.00}
]
arq = "teste_financas.csv"
salvar_lancamentos_financeiros(arq, dados_teste)
balanco = carregar_e_calcular_saldo(arq)
print(f"📊 Receitas: R$ {balanco['total_receitas']:.2f} | Despesas: R$ {balanco['total_despesas']:.2f} | Saldo: R$ {balanco['saldo_liquido']:.2f}")

if os.path.exists(arq):
    os.remove(arq)
````
