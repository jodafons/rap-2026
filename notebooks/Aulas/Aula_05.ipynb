## Introdução a Strings: Criação e Indexação

Uma string é uma sequência de caracteres. Em Python, declaramos strings usando aspas simples (`'...'`), aspas duplas (`"..."`) ou aspas triplas (três aspas simples ou três aspas duplas, úteis para strings de múltiplas linhas).

**Importante:** Strings em Python são **imutáveis**. Isso significa que, uma vez criada, você não pode alterar um caractere individual da string diretamente (ex: tentar fazer `texto[0] = "A"` causará um erro).

---

### Strings como "Arrays" de Caracteres

Para compreender a indexação de strings, é fundamental deixar claro ao aluno que **uma string se comporta exatamente como um array (vetor) de caracteres**. Cada caractere ocupa uma posição específica na memória e pode ser acessado individualmente através de seu índice (index).

Veja o exemplo prático com a palavra **"SONAR"** mapeada no sistema de coordenadas de índices:

| Índice Positivo (Esquerda para Direita) | `0` | `1` | `2` | `3` | `4` |
| :--- | :---: | :---: | :---: | :---: | :---: |
| **Caractere** | **S** | **O** | **N** | **A** | **R** |
| **Índice Negativo (Direita para Esquerda)** | `-5` | `-4` | `-3` | `-2` | `-1` |

**Conceitos-chave para o aluno:**
1. A primeira letra da palavra `"SONAR"` está na posição `0`.
2. O índice **positivo** vai de `0` até o `comprimento da palavra - 1` (de esquerda para a direita).
3. O índice **negativo** vai de `-1` (último caractere) até `-comprimento da palavra` (de direita para a esquerda). É ideal para ler dados do fim para o começo!

## Métodos Essenciais de Modificação e Busca

O Python possui uma vasta coleção de métodos internos para strings. Veja a tabela resumo a seguir com os principais métodos que estudaremos nesta seção:

| Método | Descrição | Exemplo |
| :--- | :--- | :--- |
| `len(string)` | Retorna a quantidade de caracteres da string (função built-in). | `len("Sol") -> 3` |
| `.upper()` | Converte todos os caracteres para maiúsculas. | `"radar".upper() -> "RADAR"` |
| `.lower()` | Converte todos os caracteres para minúsculas. | `"SOM".lower() -> "som"` |
| `.title()` | Converte a primeira letra de cada palavra para maiúscula. | `"capitão lucas".title() -> "Capitão Lucas"` |
| `.strip()` | Remove espaços em branco no início e no fim da string. | `"  log  ".strip() -> "log"` |
| `.replace(antigo, novo)` | Substitui todas as ocorrências de um termo por outro. | `"A1".replace("A", "B") -> "B1"` |
| `.split(delimitador)` | Divide a string em uma lista de partes usando o delimitador. | `"10;20".split(";") -> ["10", "20"]` |
| `.join(lista)` | Une elementos de uma lista usando a string como separador. | `"-".join(["A", "B"]) -> "A-B"` |
| `.startswith(prefixo)` | Verifica se a string inicia com o prefixo (retorna True/False). | `"CIAA-9".startswith("CIAA") -> True` |
| `.endswith(sufixo)` | Verifica se a string termina com o sufixo (retorna True/False). | `"sonar.log".endswith(".log") -> True` |
| `.find(termo)` | Retorna a posição (índice) da primeira ocorrência do termo ou -1. | `"Alvo".find("v") -> 2` |

### Exemplo 1: Decodificador e Validador de Sinal de Rádio

Escreva uma função `processar_sinal(sinal)` que:
1. Remova espaços em branco nas extremidades da string.
2. Converta toda a string para letras maiúsculas.
3. Verifique se o sinal começa com o prefixo `"SINAL_VALIDO_CIAA"`.
4. Se começar, extraia o código de segurança (tudo que vem após `"SINAL_VALIDO_CIAA-"`) e retorne a mensagem: `"SINAL APROVADO! Código de segurança: [CODIGO]"`.
5. Caso contrário, retorne: `"SINAL ANORMAL: Acesso negado!"`.

```python
# [Resposta]

sinal_recebido = "   sinal_valido_CIAAX-920-OK   "

def processar_sinal(sinal):

    print(f"Mensagem original: '{sinal}'")
    mensagem = sinal.strip()
    print(f"Mensagem após o strip: '{mensagem}'")

    mensagem = mensagem.upper()
    print(f"Mensagem após o upper: {mensagem}")

    if mensagem.startswith("SINAL_VALIDO_CIAA-"):

        codes = mensagem.split("-")
        print(codes)
        codigo = codes[1]
        status = codes[2]
        print(f"Código {codigo} : {status}")
        return f"SINAL APROVADO! Código de segurança: {codigo}"
    else:
        return "SINAL ANORMAL: Acesso negado!"

    
#processar_sinal(sinal_recebido)
resultado_sinal = processar_sinal(sinal_recebido)
print(f"Resultado: {resultado_sinal}")
```

### Exemplo 2: Parser de Telemetria de Sensores

Escreva uma função `parse_telemetria(dado_bruto)` que receba uma linha de dados no formato `"sensor_nome | valor_medido | unidade"` e:
1. Divida os elementos pela barra vertical `"|"`.
2. Remova os espaços em branco extras das extremidades de cada parte.
3. Converta o nome do sensor para letras maiúsculas e o valor medido para `float`.
4. Retorne a string formatada: `"Sensor [NOME_SENSOR] reportou leitura de [VALOR] [UNIDADE]."`

```python
# [Resposta]

telemetria_bruta = "profundidade_casco | 420.5 | metros"

def parse_telemetria(dado_bruto):

    mensagem_final = "Sensor [{SENSOR}] reportou leitura de [{VALOR}] [{UNIDADE}]"
    mensagem = dado_bruto.split("|")
    nome_sensor = mensagem[0].strip()
    valor = float(mensagem[1].strip())
    unidade = mensagem[2].strip()
    #print(f"'{nome_sensor}' , '{valor}', '{unidade}'")
    #mensagem_final = f"Sensor [{nome_sensor}] reportou leitura de [{valor}] [{unidade}]"
    return mensagem_final.format(SENSOR=nome_sensor, VALOR=valor, UNIDADE=unidade)


relatorio_sensor = parse_telemetria(telemetria_bruta)
print(relatorio_sensor)
```

### Exemplo 4: Filtro de Mensagens Hostis (Censura de Canal de Rádio)

**Objetivo**: Identificar e censurar termos proibidos em uma transmissão de rádio.

**Instruções**:
Escreva a função `censurar_transmissao(mensagem, termo_proibido)` que:
1. Verifique se o `termo_proibido` está contido na `mensagem` (faça a verificação de forma insensível a maiúsculas/minúsculas usando `.lower()`).
2. Se estiver presente, substitua todas as ocorrências do `termo_proibido` pela string `"CENSURADO"` na mensagem original.
3. Retorne a mensagem tratada.

*Dica: Para fazer a substituição insensível a maiúsculas/minúsculas, você pode encontrar o índice do termo na string em minúsculo e fatiar a string original, ou fazer um loop substituindo as ocorrências.*

```python
# [Resposta]

def censurar_transmissao(mensagem, lista_de_termos_proibidos, termo_substituido = "CENSURADO"):

    output = mensagem.strip()
    output = output.lower()
    for termo_proibido in lista_de_termos_proibidos:
        output = output.replace( termo_proibido, termo_substituido )
    return output

# Testando a função
msg_original = "Alerta: O reator HOSTIL foi detectado na área norte."
termos = ["hostil", "norte"]

msg_censurada = censurar_transmissao(msg_original, termos)
print("Mensagem filtrada:", msg_censurada)
# Resultado esperado: Alerta: O reator CENSURADO foi detectado na área norte.


msg_censurada = censurar_transmissao(msg_original, termos, "XXX")
print("Mensagem filtrada:", msg_censurada)
```

### Exemplo 5: Analisador de Logs de Erro

**Objetivo**: Filtrar e retornar apenas linhas de log que contenham erros ou alertas críticos.

**Instruções**:
Escreva a função `extrair_erros(logs_brutos)` que:
1. Receba uma string contendo vários logs separados por quebras de linha (`\n`).
2. Use `.split("\n")` para dividir a string em uma lista de logs individuais.
3. Crie uma lista vazia chamada `erros_detectados`.
4. Itere sobre a lista de logs e verifique se o log começa com a tag `"[ERRO]"` ou se contém a palavra `"CRÍTICO"`.
5. Se for um log de erro ou crítico, adicione-o à lista `erros_detectados`.
6. Retorne a lista de erros.

```python
# [Resposta]

def extrair_erros(logs_brutos):
    # Escreva sua lógica aqui!
    erros_detectados = []
    #logs = logs_brutos.split("\n")

    for log in logs_brutos.split("\n"):
        if log.startswith( "[ERRO]" ) or ("CRÍTICO" in log):   
            erros_detectados.append( log )
        
    return erros_detectados    

# Logs capturados do console central
historico_logs = """[OK] Sistema de ventilação operando normalmente.
[ERRO] Falha de comunicação com o sonar ativo no setor 2.
[ALERTA] Temperatura do gerador auxiliar em 75°C.
[OK] Níveis de oxigênio normais.
[LOG] Teste de sistema concluído. Alerta de reator CRÍTICO ignorado."""

lista_erros = extrair_erros(historico_logs)
#print(lista_erros)

print("Logs de erro e alertas críticos encontrados:")
for erro in lista_erros:
    print("-", erro)
# Resultado esperado:
# - [ERRO] Falha de comunicação com o sonar ativo no setor 2.
# - [LOG] Teste de sistema concluído. Alerta de reator CRÍTICO ignorado.
```

### Exemplo 6: Decodificador de Telemetria de Alvos (Desafio Tático)

**Objetivo**: Processar uma transmissão de sonar codificada que contém informações de coordenadas e tipo de alvo separadas por delimitadores especiais.

**História e Enredo**:
O receptor de hidrofones captou um sinal de alvo codificado na seguinte string:
`"ALVO:submarino#LATITUDE:-23.1209#LONGITUDE:-44.0298"`

Você deve escrever uma função para ler esse sinal bruto e extrair o tipo de alvo (em letras maiúsculas) e as coordenadas formatadas.

**Instruções**:
Escreva a função `decodificar_telemetria_alvo(sinal_bruto)` que:
1. Use `.split("#")` para dividir o sinal bruto nas três partes: alvo, latitude e longitude.
2. Extraia o valor de cada parte removendo o prefixo correspondente (ex: para `"ALVO:submarino"`, remova `"ALVO:"` e converta para maiúsculo).
3. Retorne uma string formatada exatamente no padrão:
   `"ALVO DETECTADO: [TIPO_MAIUSCULO] | COORDENADAS: [LATITUDE], [LONGITUDE]"`

```python
# [Resposta]

def decodificar_telemetria_alvo(sinal_bruto):


    mensagem_final = "ALVO DETECTADO: {ALVO} | COORDENADAS: {LATITUDE}, {LONGITUDE}"
    
    print(sinal_bruto)
    mensagem = sinal_bruto.split("#")

    #alvo = mensagem[0]
    #alvo = alvo.split(":")
    #alvo = alvo[1]
    #alvo = alvo.upper()
    #print(alvo)

    # ["ALVO", "contatorpedeiro"][1]
    alvo = mensagem[0].split(":")[1].upper()
    
    latitude  = float( mensagem[1].split(":")[1] )
    longitude = float(  mensagem[2].split(":")[1] )


    return mensagem_final.format( ALVO=alvo, LATITUDE=latitude, LONGITUDE=longitude)

# Sinal bruto interceptado pelo sonar passivo
sinal_interceptado = "ALVO:contratorpedeiro#LATITUDE:-22.9510#LONGITUDE:-43.1820"

resultado_telemetria = decodificar_telemetria_alvo(sinal_interceptado)
print(resultado_telemetria)
# Resultado esperado: ALVO DETECTADO: CONTRATORPEDEIRO | COORDENADAS: -22.9510, -43.1820
```
