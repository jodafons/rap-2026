## 1. Por que usar o NumPy?

No Python padrão, as listas são coleções de ponteiros para objetos armazenados dispersos na memória. Isso permite que uma lista armazene tipos de dados mistos (como inteiros, strings e listas juntas), mas adiciona uma grande sobrecarga de processamento.

O NumPy armazena dados em blocos de memória **contíguos** e exige que todos os elementos de um array tenham o **mesmo tipo** de dado (geralmente numérico). Isso permite:
1. **Velocidade (C-Speed)**: Operações internas são escritas em C e rodam muito mais rápido.
2. **Economia de Memória**: Ocupa uma fração do espaço de uma lista tradicional.
3. **Vetorização**: Executa loops internos implícitos, dispensando o uso de laços `for` explícitos em Python para operações matemáticas.

### Convenção de Importação
Por convenção, sempre importamos a biblioteca usando a abreviação `np`:


````python
import numpy as np

# Verificando a versão instalada no contêiner
print("Versão do NumPy:", np.__version__)
````

## 2. Criando Arrays (ndarrays)

Podemos criar arrays a partir de listas comuns, ou inicializá-los diretamente com valores específicos usando funções utilitárias.


````python
# 1. Criando a partir de uma lista comum
lista_leitura = [12.4, 15.2, 9.8, 14.7]
array_leitura = np.array(lista_leitura)
print("Array 1D:", array_leitura)

# 2. Inicializando arrays preenchidos com zeros ou uns
zeros_array = np.zeros(5) # 5 elementos iguais a 0.0
ones_array = np.ones((3, 4)) # Matriz 3x4 preenchida com 1.0
print("Zeros:", zeros_array)
print("Ones (3x4):\n", ones_array)

# 3. Criando sequências numéricas
sequencia = np.arange(0, 10, 2) # De 0 a 10 (exclusivo), de 2 em 2
grades_distancia = np.linspace(0, 100, 5) # 5 números linearmente espaçados entre 0 e 100 (inclusivo)
print("Arange:", sequencia)
print("Linspace:", grades_distancia)

# 4. Criando dados randômicos (simulando ruído de sonar)
ruido_radar = np.random.rand(3, 3) # Matriz 3x3 com números entre 0 e 1
print("Ruído Simulado (3x3):\n", ruido_radar)
````

## 3. Propriedades e Atributos dos Arrays

Cada objeto array do NumPy expõe propriedades importantes sobre sua forma e tipo de dados:


````python
painel_sensores = np.ones((4, 5))

print("Propriedades do array 'painel_sensores':")
print("- Número de Dimensões (.ndim):", painel_sensores.ndim)
print("- Formato/Shape (.shape):", painel_sensores.shape) # Retorna uma tupla (linhas, colunas)
print("- Número total de elementos (.size):", painel_sensores.size)
print("- Tipo dos dados (.dtype):", painel_sensores.dtype)
````

## 4. Indexação e Fatiamento (Slicing)

Acessar elementos no NumPy funciona como nas listas Python, mas estende-se para múltiplos eixos de forma muito elegante.


````python
# Matriz 2D simulando um grid de leituras de sonar (Linhas = Sonares, Colunas = Ângulos)
grade_sonar = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Acessando um elemento específico: matriz[linha, coluna]
print("Elemento central (linha 1, coluna 1):", grade_sonar[1, 1])
print("Elemento canto superior direito (linha 0, coluna 2):", grade_sonar[0, 2])

# Fatiamento: matriz[linha_inicio:linha_fim, coluna_inicio:coluna_fim]
print("\nPrimeira linha inteira:", grade_sonar[0, :])
print("Segunda coluna inteira:", grade_sonar[:, 1])
print("Submatriz 2x2 superior esquerda:\n", grade_sonar[0:2, 0:2])
````

### Indexação Booleana (Máscaras)

Podemos filtrar dados de um array aplicando condições lógicas diretas. Isso cria uma "máscara booleana" (um array de True/False) que usamos para extrair apenas os elementos desejados.


````python
sinais_sonar = np.array([12, 85, 45, 93, 20, 77])

# Criar máscara para sinais fortes (maiores que 50)
mascara = sinais_sonar > 50
print("Máscara Booleana:", mascara)

# Filtrar o array usando a máscara
sinais_fortes = sinais_sonar[mascara]
print("Sinais filtrados (somente > 50):", sinais_fortes)
````

## 5. Vetorização e Operações Elemento a Elemento

No NumPy, operações matemáticas comuns como adição, subtração, multiplicação e divisão são aplicadas automaticamente em todos os elementos individuais do array, sem a necessidade de loops `for`.


````python
sinal_a = np.array([1, 2, 3])
sinal_b = np.array([10, 20, 30])

# Soma elemento a elemento
print("Soma:", sinal_a + sinal_b)

# Multiplicação por escalar (amplificação do sinal)
print("Sinal Amplificado:", sinal_a * 5)

# Funções matemáticas prontas aplicadas ao array
valores_radianos = np.array([0, np.pi/2, np.pi])
print("Seno dos valores:", np.sin(valores_radianos))
````

## 6. Funções de Agregação e Eixos (Axes)

O NumPy oferece funções rápidas para calcular estatísticas agregadas. Em matrizes bidimensionais, podemos calcular essas estatísticas para a matriz inteira, por colunas, ou por linhas usando o parâmetro `axis`:

* `axis=0`: Executa a operação **verticalmente** (calcula por coluna, esmagando as linhas).
* `axis=1`: Executa a operação **horizontalmente** (calcula por linha, esmagando as colunas).


````python
# Matriz de temperaturas de 2 reatores monitorados em 3 instantes de tempo
# Linha 0: Reator A, Linha 1: Reator B
temperaturas = np.array([
    [250.0, 280.0, 275.0],  # Reator A
    [290.0, 310.0, 285.0]   # Reator B
])

# 1. Média global da matriz inteira
print("Temperatura média geral:", np.mean(temperaturas))

# 2. Temperatura máxima de cada reator (ao longo das colunas, por linha)
print("Máxima por Reator (axis=1):", np.max(temperaturas, axis=1))

# 3. Soma das temperaturas por instante de tempo (ao longo das linhas, por coluna)
print("Soma por instante de tempo (axis=0):", np.sum(temperaturas, axis=0))
````

## 7. Álgebra Linear com NumPy

O NumPy oferece um conjunto de funções de álgebra linear de altíssima performance no submódulo `np.linalg` e através de operadores nativos do próprio array.

### 7.1. Produto Escalar (Dot Product)
O produto escalar multiplica os elementos correspondentes de dois vetores e soma os resultados. É muito utilizado no cálculo de ponderação de sinais de sensores.
Podemos calculá-lo com `np.dot(a, b)` ou com o operador `@`:


````python
vetor_a = np.array([1, 2, 3])
vetor_b = np.array([4, 5, 6])

# Cálculo do produto escalar (1*4 + 2*5 + 3*6 = 4 + 10 + 18 = 32)
produto_escalar_1 = np.dot(vetor_a, vetor_b)
produto_escalar_2 = vetor_a @ vetor_b

print("Produto Escalar (np.dot):", produto_escalar_1)
print("Produto Escalar (operador @):", produto_escalar_2)
````

### 7.2. Multiplicação de Matrizes (Matrix Multiplication)
Diferente da multiplicação comum elemento a elemento (`*`), a multiplicação de matrizes segue a regra de produto de linhas por colunas.
Usamos `np.matmul(A, B)` ou o operador `@`:


````python
matriz_a = np.array([[1, 2], [3, 4]])
matriz_b = np.array([[5, 6], [7, 8]])

# Multiplicação de matrizes
resultado_matmul = np.matmul(matriz_a, matriz_b)
resultado_operador = matriz_a @ matriz_b

print("Matriz A @ Matriz B (np.matmul):\n", resultado_matmul)
print("Matriz A @ Matriz B (operador @):\n", resultado_operador)

print("\nPara comparação, multiplicação comum elemento a elemento (*):\n", matriz_a * matriz_b)
````

### 7.3. Transposição de Matriz (Transpose)
Transpor uma matriz consiste em trocar suas linhas por colunas. No NumPy, basta usar a propriedade `.T` ou a função `np.transpose()`:


````python
matriz_original = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matriz Original (2x3):\n", matriz_original)
print("Matriz Transposta (3x2) usando .T:\n", matriz_original.T)
````

### 7.4. Inversão de Matrizes
A inversa de uma matriz quadrada $A$ é a matriz $A^{-1}$ tal que $A \cdot A^{-1} = I$ (onde $I$ é a matriz identidade).
Podemos calcular usando `np.linalg.inv()`:


````python
matriz_quadrada = np.array([
    [1, 2],
    [3, 4]
])

# Inversão
matriz_inversa = np.linalg.inv(matriz_quadrada)
print("Matriz Inversa:\n", matriz_inversa)

# Verificando: Matriz * Inversa deve resultar na Matriz Identidade
print("\nVerificação (A @ A_inv):\n", np.round(matriz_quadrada @ matriz_inversa))
````

### 7.5. Resolução de Sistemas Lineares
Muitos problemas de engenharia e física consistem em resolver equações lineares do tipo $A \cdot x = b$, onde queremos descobrir o vetor $x$.
O NumPy possui a função extremamente otimizada `np.linalg.solve(A, b)` para resolver esses sistemas.

**Exemplo Prático (Triangulação de Sonar no CIAA-LPS):**
Deseja-se localizar as coordenadas $(x, y)$ de um alvo subaquático a partir de duas leituras de sensores de sonar. Chegamos ao seguinte sistema linear:
$$1x + 2y = 5$$
$$3x + 4y = 11$$

Vamos resolvê-lo:


````python
# Matriz dos coeficientes (A)
A = np.array([
    [1, 2],
    [3, 4]
])

# Vetor dos resultados (b)
b = np.array([5, 11])

# Resolvendo Ax = b para encontrar x
x = np.linalg.solve(A, b)
print("Coordenadas do Alvo (x, y):", x)
````
