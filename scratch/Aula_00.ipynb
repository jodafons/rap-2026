## 2. O que é Docker e Contêineres?

Imagine que você precisa enviar mercadorias por navios de carga. Se você colocar roupas, grãos, computadores e carros soltos no navio, as mercadorias vão se misturar, quebrar e estragar. A solução do mundo real foi criar **Containers de Carga Padronizados**.

No desenvolvimento de software, o **Docker** faz exatamente isso! Ele é uma plataforma de código aberto que permite empacotar uma aplicação (com todo o seu código, bibliotecas, Python e dependências) dentro de uma unidade isolada chamada **Contêiner**.

### O Problema do "Funciona na Minha Máquina":
> [!IMPORTANT]> Sem o Docker, se um aluno usa Windows, outro usa macOS e outro Linux, cada um teria versões de Python, pacotes e caminhos de arquivos diferentes, gerando bugs difíceis de resolver.> > Com o Docker, todo o nosso ambiente de desenvolvimento (incluindo o Jupyter Lab e todas as bibliotecas necessárias) roda de forma idêntica e isolada dentro de um contêiner baseado em Linux, independentemente do sistema operacional da sua máquina física.
### Diferença entre Contêineres e Máquinas Virtuais (VMs):
* **Máquinas Virtuais (VMs)**: Emulam um hardware físico inteiro e exigem um Sistema Operacional completo e pesado para rodar cada ambiente, consumindo muito processamento e memória.
* **Contêineres Docker**: Compartilham o Kernel do sistema operacional do computador hospedeiro. Eles não precisam simular um hardware inteiro, o que os torna incrivelmente leves, rápidos e com inicialização quase instantânea.


## 3. O que é Docker Desktop?

O **Docker Desktop** é uma aplicação gráfica instalada no seu computador (Windows ou macOS) que serve para gerenciar de forma amigável tudo o que acontece no Docker.

### O que ele faz por nós?
* **Painel de Controle Visível**: Permite ver quais contêineres estão rodando, pará-los, iniciá-los ou excluí-los com apenas um clique.
* **Visualização de Logs**: Facilita ver as mensagens de saída de texto geradas pelos aplicativos que rodam dentro do contêiner.
* **Gerenciamento de Imagens e Volumes**: Permite baixar imagens prontas da internet e controlar pastas do seu computador físico que são compartilhadas com o contêiner (Volumes).


## 4. O que é Jupyter Lab?

O **Jupyter Lab** é a interface web interativa onde estamos trabalhando neste momento. Ele roda dentro do nosso contêiner Docker e exibe páginas interativas diretamente no navegador do seu computador.

### O que são Notebooks (`.ipynb`)?
Um arquivo de notebook (com a extensão `.ipynb`) é um documento interativo que mescla textos explicativos, fórmulas matemáticas, gráficos e códigos de computador executáveis no mesmo lugar. Ele é composto por **Células**.

Existem dois tipos principais de células que você usará neste curso:
1. **Células de Markdown (Texto)**: Servem para explicações, formatações de títulos, listas e fórmulas. Esta célula que você está lendo agora é uma célula de Markdown.
2. **Células de Código**: Contêm comandos em Python que podem ser enviados para execução. Abaixo de cada célula de código, é exibido o resultado impresso correspondente.


## 5. Guia Rápido de Atalhos do Jupyter Lab

Para ser produtivo no Jupyter Lab, você deve dominar alguns atalhos de teclado. O Jupyter tem dois modos de operação:
* **Modo de Edição (borda verde/azul na célula)**: Quando você está digitando texto dentro de uma célula. (Ative pressionando `Enter` na célula).
* **Modo de Comando (borda cinza na célula)**: Quando você está navegando pelas células sem digitar nelas. (Ative pressionando `Esc`).

### Atalhos úteis no Modo de Comando (pressione `Esc` primeiro):
* `Shift + Enter`: Executa a célula atual e seleciona a próxima célula abaixo.
* `Ctrl + Enter`: Executa a célula atual e permanece nela.
* `A` (Above): Adiciona uma nova célula em branco **acima** da atual.
* `B` (Below): Adiciona uma nova célula em branco **abaixo** da atual.
* `M` (Markdown): Transforma a célula atual em uma célula de **Texto (Markdown)**.
* `Y` (Code): Transforma a célula atual em uma célula de **Código Python**.
* `D, D` (Pressionar a tecla D duas vezes rapidamente): Exclui a célula atual.
