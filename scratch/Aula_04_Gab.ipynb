## Introdução às Listas e Suas Principais Funções

Até agora, cada variável que declaramos armazenava apenas um único valor por vez. Em problemas reais de engenharia e análise de dados, precisamos trabalhar com conjuntos de dados (como uma série temporal de temperaturas de sensores ou nomes de oficiais de serviço).

Para isso, o Python nos fornece a estrutura de dados chamada **Lista** (`list`).

### O que é uma Lista?
Uma lista é uma coleção ordenada, mutável (pode ser alterada) e que permite elementos duplicados. Em Python, as listas são representadas por colchetes `[]` e seus elementos são separados por vírgula.

### Principais Funções e Métodos de Listas

O Python oferece diversas funções integradas e métodos internos para manipular listas. Veja a tabela resumo a seguir:

| Função / Método | Descrição | Exemplo |
| :--- | :--- | :--- |
| `len(lista)` | Retorna a quantidade de elementos na lista (função built-in). | `len([10, 20]) -> 2` |
| `.append(item)` | Adiciona um elemento ao **final** da lista. | `lista.append(50)` |
| `.extend(iteravel)` | Adiciona todos os elementos de outro iterável (como uma lista) ao final. | `lista.extend([30, 40])` |
| `.insert(indice, item)`| Insere um elemento em uma posição (índice) específica. | `lista.insert(0, 10)` |
| `.remove(item)` | Remove a primeira ocorrência do elemento especificado. | `lista.remove(20)` |
| `.pop(indice)` | Remove e retorna o item na posição especificada (por padrão, o último). | `lista.pop()` |
| `.index(item)` | Retorna o índice (posição) da primeira ocorrência do item. | `lista.index(10)` |
| `.count(item)` | Retorna a quantidade de ocorrências de um item na lista. | `lista.count(5)` |
| `.sort()` | Ordena a lista em ordem crescente (in-place). | `lista.sort()` |
| `.reverse()` | Inverte a ordem dos elementos da lista (in-place). | `lista.reverse()` |
| `.clear()` | Remove todos os elementos da lista, deixando-a vazia. | `lista.clear()` |

````python
# [Exemplo] Iteração com loops aninhados (for dentro de for)
# Vamos percorrer cada setor da nossa grade de navegação e classificar as ameaças detectadas.
# Se for maior ou igual a 4, você deve imprimir uma mensagem de ameaça detectada, caso contrário, sinal fraco.
# Use enumerate
mapa_ameacas = [
    [0, 1, 3],
    [5, 0, 2],
    [1, 0, 0]
]
for idx_linha, linha in enumerate(mapa_ameacas):
    for idx_coluna, valor in enumerate(linha):
        if valor >= 4:
            print(f"Setor ({idx_linha}, {idx_coluna}): Ameaça Crítica Detectada ({valor})!")
        else:
            print(f"Setor ({idx_linha}, {idx_coluna}): Sinal fraco.")
````

### Exemplo 1:

Uma pasta do computador contém a seguinte lista de nomes de arquivos:
```python
# lista de arquivos no diretório
arquivos = ["documento.pdf", "foto.png", "relatorio.docx", "planilha.xlsx", "video.mp4"]
```

Crie a função `buscar_arquivo` que busque por um arquivo específico (por exemplo, `relatorio.docx`). A função deve percorrer a lista item por item. Se encontrar o arquivo, deve exibir a mensagem `Arquivo encontrado na posição X!` e interromper a busca imediatamente (`break`). Caso percorra a lista inteira e não encontre, exiba `Arquivo não encontrado`.


````python
# [Resposta]
def buscar_arquivo(arquivos, nome_procurado):
    encontrado = False
    for idx, arquivo in enumerate(arquivos):
        if arquivo == nome_procurado:
            print(f"Arquivo encontrado na posição {idx}!")
            encontrado = True
            break
    if not encontrado:
        print("Arquivo não encontrado")

arquivos = ["documento.pdf", "foto.png", "relatorio.docx", "planilha.xlsx", "video.mp4"]
buscar_arquivo(arquivos, "relatorio.docx")
````

### Exemplo 2: Calcular média

Escreva a função `calcular_media` onde deve receber como entrada uma lista de numeros e retornar a média. O valor de retorno deve ser truncado com duas casas decimais usando a função `round()`.


````python
# [Resposta]
def calcular_media(valores):
    if not valores:
        return 0.0
    return round(sum(valores) / len(valores), 2)

# Conjunto de medições em decibéis captadas pelo sonar passivo
medicoes = [45.2, 48.7, 51.1, 46.3, 49.0, 50.0, 51.0]
media_db = calcular_media(medicoes)
print(f"Medições coletadas: {medicoes}")
print(f"Média do ruído de fundo: {media_db} dB")
````

### Exemplo 3:

Escreva uma função em Python que recebe por argumento, e nessa ordem, uma lista de números e um número, e retorna uma nova lista com as posições em que esse número aparece na lista passada por argumento. Se o número não existir na lista, a função deve retornar uma lista vazia.

````python
# [Resposta]
def encontrar_posicoes(lista, numero):
    posicoes = []
    for idx, item in enumerate(lista):
        if item == numero:
            posicoes.append(idx)
    return posicoes

print(encontrar_posicoes([1, 2, 3, 2, 4, 2], 2))
````

### Exemplo 4: Encode de RLE

(Inspirada na OBI 2023 – Nível Júnior, Fase 2) Em computação, compressão de arquivos é o processo de reduzir o tamanho de arquivos. Uma técnica simples, mas eficiente de compressão para alguns tipos de arquivos, é RLE, cujo nome é derivado das letras iniciais do nome em inglês, que pode ser traduzido por codificação por comprimento de repetições.

Em RLE, cada sequência de valores repetidos no arquivo é substituída pelo número de repetições seguido do valor a ser repetido. Por exemplo, a lista `[1, 1, 1, 1, 1, 1]` seria substituída por `[6, 1]`, pois há 6 números 1. Se o arquivo tem muitas sequências de valores repetidos, a técnica RLE é bem eficiente (por exemplo, considere um arquivo de imagem branco e preto em que grande parte dos valores é branco). Por outro lado, se o arquivo não possui muitas repetições, a técnica RLE pode aumentar o tamanho do arquivo, ao invés de diminuir. 

Escreva a função `encodar_rle` que recebe por argumento uma lista de números e retorna uma nova lista que representa a primeira, mas comprimida com a técnica RLE. Veja que a técnica **só considera números iguais em posições vizinhas**, se houver outro número no meio, a contagem é reiniciada. 

Por exemplo, para a entrada `[1, 1, 1, 1, 2, 1, 1, 1]`, a saída é `[4, 1, 1, 2, 3, 1]`, pois há 4 números 1 seguidos, 1 número 2 e 3 números 1. 

Para a entrada `[1, 2, 3, 4, 1]`, a saída é `[1, 1, 1, 2, 1, 3, 1, 4, 1, 1]`, pois há 1 número 1, 1 número 2, 1 número 3, 1 número 4 e mais 1 número 1. Esse seria um caso ruim para esse tipo de compressão, pois há muitas mudanças de valor. Esse tipo de compressão é boa quando há muitas repetições seguidas.

````python
# [Resposta]
def encodar_rle(entrada):
    if not entrada:
        return []
    saida = []
    contador = 1
    for i in range(1, len(entrada)):
        if entrada[i] == entrada[i - 1]:
            contador += 1
        else:
            saida.extend([contador, entrada[i - 1]])
            contador = 1
    saida.extend([contador, entrada[-1]])
    return saida

# Testando a função de controle de estoque
entrada_a = [1, 1, 1, 1, 2, 1, 1, 1]
entrada_b = [1, 2, 3, 4, 1]
saida_a = encodar_rle(entrada_a)
saida_b = encodar_rle(entrada_b)

print(f"RLE de {entrada_a} é {saida_a}")
print(f"RLE de {entrada_b} é {saida_b}")
````

### Exemplo 5: produto escalar de matrizes

Para calcular o produto escalar (dot product) de duas matrizes na mão (representadas em Python como lista de listas), nós fazemos a clássica multiplicação de matrizes: o elemento da linha i e coluna j da matriz resultado é a soma da multiplicação linha por coluna.

Para ser possível calcular o produto escalar de $A\cdot B$, o número de colunas de $A$ deve ser igual ao número de linhas de $B$.

Dadas as matrizes:
$$A = \begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{bmatrix}, \quad B = \begin{bmatrix} 7 & 8 \\ 9 & 1 \\ 2 & 3 \end{bmatrix}$$

Regra geral para encontrar cada posição $C[i][j]$: **Multiplique a Linha $i$ de $A$ pela Coluna $j$ de $B$ e some os produtos.**

* **Posição $C[0][0]$** (Linha 0 de $A \times$ Coluna 0 de $B$):
  $$(1 \cdot 7) + (2 \cdot 9) + (3 \cdot 2) = 7 + 18 + 6 = \mathbf{31}$$

* **Posição $C[0][1]$** (Linha 0 de $A \times$ Coluna 1 de $B$):
  $$(1 \cdot 8) + (2 \cdot 1) + (3 \cdot 3) = 8 + 2 + 9 = \mathbf{19}$$

* **Posição $C[1][0]$** (Linha 1 de $A \times$ Coluna 0 de $B$):
  $$(4 \cdot 7) + (5 \cdot 9) + (6 \cdot 2) = 28 + 45 + 12 = \mathbf{85}$$

* **Posição $C[1][1]$** (Linha 1 de $A \times$ Coluna 1 de $B$):
  $$(4 \cdot 8) + (5 \cdot 1) + (6 \cdot 3) = 32 + 5 + 18 = \mathbf{55}$$

**Matriz Resultante $C$:**
$$C = \begin{bmatrix} 31 & 19 \\ 85 & 55 \end{bmatrix}$$


Implementação em Python Passo a Passo
1. **Primeiro `for` ($i$):** Percorre cada linha de $A$.
2. **Segundo `for` ($j$):** Percorre cada coluna de $B$.
3. **Terceiro `for` ($k$):** Anda simultaneamente pelos elementos da linha de $A$ e coluna de $B$ realizando a multiplicação e o acúmulo da soma.



````python
# [Resposta]
def calcular_produto_escalar(A, B):
    num_linhas_A = len(A)
    num_colunas_A = len(A[0])
    num_colunas_B = len(B[0])
    
    # Cria matriz resultado C com zeros
    C = [[0] * num_colunas_B for _ in range(num_linhas_A)]
    
    for i in range(num_linhas_A):
        for j in range(num_colunas_B):
            soma = 0
            for k in range(num_colunas_A):
                soma += A[i][k] * B[k][j]
            C[i][j] = soma
    return C

A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [7, 8],
    [9, 1],
    [2, 3]
]

C = calcular_produto_escalar(A, B)
print("Resultado do produto escalar de matrizes:")
for linha in C:
    print(linha)
````

### Exemplo 6: Classificador Tático de Contatos Acústicos Multidimensionais

O computador tático do submarino detecta múltiplos contatos simultâneos na tela do sonar. Cada contato é representado por uma lista contendo três dados numéricos: `[frequencia, amplitude, tempo_eco]`, onde:
*   `frequencia`: Frequência da leitura acústica em Hz.
*   `amplitude`: Amplitude do sinal em dB.
*   `tempo_eco`: Tempo que o eco demorou para ir e voltar do alvo em segundos.

Você deve reutilizar a lógica de classificação de alvos que desenvolvemos na primeira e segunda aula, porém agora aplicando-a de forma automatizada sobre múltiplos alvos recebidos em uma lista de contatos.

**Instruções**:
1.  Reescreva a função auxiliar `calcular_distancia_aux(velocidade_som, tempo_eco)` que retorna a distância simples em metros utilizando a fórmula:
    $$\text{distancia} = \frac{\text{velocidade\_som} \times \text{tempo\_eco}}{2}$$
2.  Escreva a função principal `classificar_lista_contatos(lista_contatos, velocidade_som)` que:
    *   Crie uma lista vazia chamada `classificacoes`.
    *   Itere sobre cada contato contido na `lista_contatos` (cada contato é uma sublista contendo `[frequencia, amplitude, tempo_eco]`).
    *   Calcule a distância até o contato chamando a função `calcular_distancia_aux` com `velocidade_som` e o `tempo_eco` correspondente.
    *   Classifique o alvo seguindo as regras:
        *   Se a frequência for entre $500\text{ Hz}$ e $1500\text{ Hz}$ (inclusive) AND amplitude maior ou igual a $80\text{ dB}$: classifica como `"Submarino"`.
        *   Se a frequência for entre $20\text{ Hz}$ e $200\text{ Hz}$ (inclusive) AND amplitude maior ou igual a $60\text{ dB}$: classifica como `"Baleia"`.
        *   Caso contrário: classifica como `"Ruído"`.
    *   Adicione a string com a classificação correspondente à lista `classificacoes`.
    *   Ao final do processamento de todos os alvos, retorne a lista `classificacoes`.

````python
# [Resposta]
def calcular_distancia_aux(velocidade_som, tempo_eco):
    return (velocidade_som * tempo_eco) / 2

def classificar_lista_contatos(lista_contatos, velocidade_som):
    classificacoes = []
    for contato in lista_contatos:
        frequencia = contato[0]
        amplitude = contato[1]
        tempo_eco = contato[2]
        dist = calcular_distancia_aux(velocidade_som, tempo_eco)
        if 500 <= frequencia <= 1500 and amplitude >= 80:
            classificacoes.append("Submarino")
        elif 20 <= frequencia <= 200 and amplitude >= 60:
            classificacoes.append("Baleia")
        else:
            classificacoes.append("Ruído")
    return classificacoes

# Banco de contatos captados pelo sonar ativo
contatos_banco = [
    [850.0, 95.0, 1.4],   # Esperado: Submarino
    [50.0, 72.0, 1.2],    # Esperado: Baleia
    [1200.0, 45.0, 2.0],  # Esperado: Ruído
    [10.0, 85.0, 0.5]     # Esperado: Ruído
]
velocidade_som_agua = 1500.0 # m/s

diagnosticos = classificar_lista_contatos(contatos_banco, velocidade_som_agua)
print("Relatório de Classification Tática de Alvos:")
for i, classe in enumerate(diagnosticos):
    print(f"Contato #{i+1}: Classificado como '{classe}'")
````
